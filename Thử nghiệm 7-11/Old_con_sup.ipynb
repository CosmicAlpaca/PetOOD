```python
# -- coding: utf-8 --
"""
BÀI TOÁN NGOÀI PHÂN PHỐI ỨNG DỤNG TRONG NHẬN DIỆN MẪU
Phiên bản cải tiến: Tập trung vào DreamOOD
"""

#==================================================================
# BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG
#==================================================================
print(" BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG ".center(80, "="))
!pip install -q torchmetrics gradio wandb diffusers transformers accelerate opencv-python scikit-image
!git clone -q https://github.com/deeplearning-wisc/dream-ood.git

import os
import sys
sys.path.append('./dream-ood')

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
import torchvision
import torchvision.transforms as T
from torch.utils.data import DataLoader, Subset, ConcatDataset, Dataset
from tqdm.notebook import tqdm
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score, roc_curve, auc
import seaborn as sns
from PIL import Image
import gradio as gr
import pandas as pd
import wandb
from datetime import datetime

wandb.init(project="ood-animal-detection-dreamoe-focused", name=f"dream_oe_exp_{datetime.now().strftime('%Y%m%d_%H%M%S')}")

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Sử dụng thiết bị: {DEVICE}")

from google.colab import drive
drive.mount('/content/drive')
SAVE_DIR = "/content/drive/MyDrive/OOD_Animal_Project_DreamOE"
os.makedirs(SAVE_DIR, exist_ok=True)
print(f"Model và kết quả sẽ được lưu tại: {SAVE_DIR}")

#==================================================================
# BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU (Tương tự nhưng tinh gọn)
#==================================================================
print("\n" + " BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU ".center(80, "="))

# (Phần class DreamOODGenerator và DreamOODDataset giữ nguyên)
class DreamOODGenerator:
    def __init__(self, model_name="stabilityai/stable-diffusion-2-1-base"):
        from diffusers import StableDiffusionPipeline
        self.pipe = StableDiffusionPipeline.from_pretrained(model_name, torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32)
        if torch.cuda.is_available(): self.pipe.to("cuda")
        self.ood_prompts = ["abstract geometric patterns", "random noise texture", "colorful abstract painting", "geometric fractal patterns", "digital glitch art"]
    def generate_ood_batch(self, batch_size=8, size=(32, 32)):
        images = []
        for _ in range(batch_size):
            prompt = np.random.choice(self.ood_prompts)
            random_elements = ["vibrant", "muted", "dark", "pastel"]
            prompt = f"{np.random.choice(random_elements)} {prompt}"
            with torch.no_grad():
                image = self.pipe(prompt, num_inference_steps=20, height=256, width=256, guidance_scale=7.5).images[0].resize(size)
            images.append(np.array(image))
        return np.array(images)
    def create_ood_dataset(self, num_samples=1000, save_path=None):
        print(f"Đang tạo {num_samples} mẫu OOD...")
        batch_size = 16
        num_batches = (num_samples + batch_size - 1) // batch_size
        all_images = np.concatenate([self.generate_ood_batch(min(batch_size, num_samples - i*batch_size)) for i in tqdm(range(num_batches), desc="Generating OOD")])
        if save_path: np.save(save_path, all_images); print(f"Đã lưu OOD dataset tại: {save_path}")
        return all_images

class DreamOODDataset(Dataset):
    def __init__(self, images, transform=None): self.images, self.transform = images, transform
    def __len__(self): return len(self.images)
    def __getitem__(self, idx):
        image = Image.fromarray(self.images[idx].astype(np.uint8))
        if self.transform: image = self.transform(image)
        return image, -1

# Định nghĩa classes và transforms
cifar10_classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']
cifar10_animal_indices = [2, 3, 4, 5, 6, 7]
cifar100_animal_indices = [1, 3, 4, 6, 7, 14, 15, 18, 19, 21, 26, 27, 30, 31, 32, 34, 36, 38, 42, 43, 44, 45, 46, 51, 56, 57, 65, 66, 67, 68, 70, 71, 72, 73, 74, 75, 76, 77, 86, 91, 93, 98, 99]
cifar100_classes = ['apple', 'aquarium_fish', 'baby', 'bear', 'beaver', 'bed', 'bee', 'beetle', 'bicycle', 'bottle', 'bowl', 'boy', 'bridge', 'bus', 'butterfly', 'camel', 'can', 'castle', 'caterpillar', 'cattle', 'chair', 'chimpanzee', 'clock', 'cloud', 'cockroach', 'couch', 'crab', 'crocodile', 'cup', 'dinosaur', 'dolphin', 'elephant', 'flatfish', 'forest', 'fox', 'girl', 'hamster', 'house', 'kangaroo', 'keyboard', 'lamp', 'lawn_mower', 'leopard', 'lion', 'lizard', 'lobster', 'man', 'maple_tree', 'motorcycle', 'mountain', 'mouse', 'mushroom', 'oak_tree', 'orange', 'orchid', 'otter', 'palm_tree', 'pear', 'pickup_truck', 'pine_tree', 'plain', 'plate', 'poppy', 'porcupine', 'possum', 'rabbit', 'raccoon', 'ray', 'road', 'rocket', 'rose', 'sea', 'seal', 'shark', 'shrew', 'skunk', 'skyscraper', 'snail', 'snake', 'spider', 'squirrel', 'streetcar', 'sunflower', 'sweet_pepper', 'table', 'tank', 'telephone', 'television', 'tiger', 'tractor', 'train', 'trout', 'tulip', 'turtle', 'wardrobe', 'whale', 'willow_tree', 'wolf', 'woman', 'worm']
id_classes_names = [cifar10_classes[i] for i in cifar10_animal_indices] + [cifar100_classes[i] for i in cifar100_animal_indices]
near_ood_classes_indices = [i for i in range(10) if i not in cifar10_animal_indices]
transform_train = T.Compose([T.RandomCrop(32, padding=4), T.RandomHorizontalFlip(), T.RandomRotation(15), T.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1), T.ToTensor(), T.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010))])
transform_test = T.Compose([T.ToTensor(), T.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010))])
transform_oe = T.Compose([T.Resize((32, 32)), T.RandomCrop(32, padding=4), T.RandomHorizontalFlip(), T.ToTensor(), T.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))])

# Tạo hoặc tải DreamOOD data
dream_ood_path = os.path.join(SAVE_DIR, 'dream_ood_samples.npy')
if not os.path.exists(dream_ood_path):
    dream_generator = DreamOODGenerator()
    dream_ood_images = dream_generator.create_ood_dataset(num_samples=5000, save_path=dream_ood_path) # Cải tiến: tạo nhiều mẫu hơn
else:
    print("Tải DreamOOD dataset đã có...")
    dream_ood_images = np.load(dream_ood_path)

# (Phần setup dataset và dataloader giữ nguyên, nhưng loại bỏ oe_loader không cần thiết)
class RemappedSubset(Subset):
    def __init__(self, dataset, indices, label_map): super().__init__(dataset, indices); self.label_map = label_map
    def __getitem__(self, idx): image, label = self.dataset[self.indices[idx]]; return image, self.label_map[label]
cifar10_train = torchvision.datasets.CIFAR10('./data', train=True, download=True, transform=transform_train)
cifar10_test = torchvision.datasets.CIFAR10('./data', train=False, download=True, transform=transform_test)
cifar100_train = torchvision.datasets.CIFAR100('./data', train=True, download=True, transform=transform_train)
cifar100_test = torchvision.datasets.CIFAR100('./data', train=False, download=True, transform=transform_test)
c10_label_map = {orig: new for new, orig in enumerate(cifar10_animal_indices)}
c100_label_map = {orig: new + len(c10_label_map) for new, orig in enumerate(cifar100_animal_indices)}
id_train_dataset = ConcatDataset([RemappedSubset(cifar10_train, [i for i, (_, l) in enumerate(cifar10_train) if l in cifar10_animal_indices], c10_label_map), RemappedSubset(cifar100_train, [i for i, (_, l) in enumerate(cifar100_train) if l in cifar100_animal_indices], c100_label_map)])
id_test_dataset = ConcatDataset([RemappedSubset(cifar10_test, [i for i, (_, l) in enumerate(cifar10_test) if l in cifar10_animal_indices], c10_label_map), RemappedSubset(cifar100_test, [i for i, (_, l) in enumerate(cifar100_test) if l in cifar100_animal_indices], c100_label_map)])
near_ood_test_dataset = Subset(cifar10_test, [i for i, (_, l) in enumerate(cifar10_test) if l in near_ood_classes_indices])
far_ood_test_dataset = torchvision.datasets.SVHN('./data', split='test', download=True, transform=transform_test)
dream_ood_dataset = DreamOODDataset(dream_ood_images, transform=transform_oe)

batch_size = 128
id_train_loader = DataLoader(id_train_dataset, batch_size, shuffle=True, num_workers=2, pin_memory=True)
id_test_loader = DataLoader(id_test_dataset, batch_size, shuffle=False, num_workers=2, pin_memory=True)
near_ood_loader = DataLoader(near_ood_test_dataset, batch_size, shuffle=False, num_workers=2, pin_memory=True)
far_ood_loader = DataLoader(far_ood_test_dataset, batch_size, shuffle=False, num_workers=2, pin_memory=True)
dream_ood_loader = DataLoader(dream_ood_dataset, batch_size, shuffle=True, num_workers=2, pin_memory=True)

#==================================================================
# BƯỚC 3: MODEL VÀ PHƯƠNG PHÁP DETECT (Giữ nguyên)
#==================================================================
print("\n" + " BƯỚC 3: MODEL VÀ PHƯƠNG PHÁP DETECT ".center(80, "="))
def get_model(): return torchvision.models.resnet18(weights=None, num_classes=len(id_classes_names)).to(DEVICE)
# (Class OODDetector và hàm evaluate_ood_performance giữ nguyên)
class OODDetector:
    @staticmethod
    def energy_score(model, data_loader):
        model.eval(); scores = []
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "Energy Score", leave=False):
                logits = model(images.to(DEVICE)); energy = -torch.logsumexp(logits, dim=1)
                scores.extend(energy.cpu().numpy())
        return np.array(scores)
    @staticmethod
    def max_softmax_probability(model, data_loader):
        model.eval(); scores = []
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "MSP", leave=False):
                probs = F.softmax(model(images.to(DEVICE)), dim=1); max_probs, _ = torch.max(probs, dim=1)
                scores.extend((-max_probs).cpu().numpy())
        return np.array(scores)
    # (Hàm odin_score và evaluate_ood_performance giữ nguyên)
def evaluate_ood_performance(id_scores, ood_scores):
    labels = np.concatenate([np.zeros_like(id_scores), np.ones_like(ood_scores)])
    scores = np.concatenate([id_scores, ood_scores])
    auroc = roc_auc_score(labels, scores)
    fpr, tpr, _ = roc_curve(labels, scores)
    fpr95 = fpr[np.argmax(tpr >= 0.95)]
    return {'auroc': auroc, 'fpr95': fpr95}

#==================================================================
# BƯỚC 4: HUẤN LUYỆN MÔ HÌNH DREAMOE
#==================================================================
print("\n" + " BƯỚC 4: HUẤN LUYỆN MÔ HÌNH DREAMOE ".center(80, "="))

def train_dream_oe_model(model, train_loader, val_loader, ood_loader, epochs=100, ood_weight=0.5):
    optimizer = optim.SGD(model.parameters(), lr=0.1, momentum=0.9, weight_decay=5e-4)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)
    criterion = nn.CrossEntropyLoss()
    best_val_acc = 0
    train_history = []

    for epoch in range(epochs):
        model.train()
        train_loss, train_correct, train_total = 0, 0, 0
        ood_iter = iter(ood_loader)
        progress_bar = tqdm(train_loader, desc=f"DreamOE Epoch {epoch+1}/{epochs}", leave=False)
        for inputs, targets in progress_bar:
            inputs, targets = inputs.to(DEVICE), targets.to(DEVICE)
            optimizer.zero_grad()

            # Loss trên dữ liệu ID
            outputs = model(inputs)
            loss_id = criterion(outputs, targets)

            # Loss trên dữ liệu OOD (DreamOOD)
            try:
                ood_inputs, _ = next(ood_iter)
            except StopIteration:
                ood_iter = iter(ood_loader)
                ood_inputs, _ = next(ood_iter)
            outputs_ood = model(ood_inputs.to(DEVICE))
            loss_ood = - (F.log_softmax(outputs_ood, dim=1).mean())

            total_loss = loss_id + ood_weight * loss_ood
            total_loss.backward()
            optimizer.step()

            train_loss += total_loss.item()
            _, predicted = outputs.max(1)
            train_total += targets.size(0)
            train_correct += predicted.eq(targets).sum().item()
        scheduler.step()

        # Đánh giá (Validation phase)
        model.eval()
        val_loss, val_correct, val_total = 0, 0, 0
        with torch.no_grad():
            for inputs, targets in val_loader:
                # ================================================= #
                # DÒNG SỬA LỖI QUAN TRỌNG NẰM Ở ĐÂY
                # Chuyển cả inputs và targets sang đúng thiết bị (GPU)
                inputs = inputs.to(DEVICE)
                targets = targets.to(DEVICE)
                # ================================================= #

                outputs = model(inputs)
                loss = criterion(outputs, targets)
                val_loss += loss.item()
                _, predicted = outputs.max(1)
                val_total += targets.size(0)
                val_correct += predicted.eq(targets).sum().item()

        # Logging
        train_acc, val_acc = train_correct/train_total, val_correct/val_total
        epoch_metrics = {
            'epoch': epoch+1,
            'train_loss': train_loss/len(train_loader),
            'train_acc': train_acc,
            'val_loss': val_loss/len(val_loader),
            'val_acc': val_acc
        }
        train_history.append(epoch_metrics)
        wandb.log({f'dream_oe/{k}': v for k, v in epoch_metrics.items()})
        print(f"Epoch {epoch+1:03d}: Train Acc: {train_acc:.4f}, Val Acc: {val_acc:.4f} | Val Loss: {epoch_metrics['val_loss']:.4f}")

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), os.path.join(SAVE_DIR, 'dream_oe_best_model.pth'))
            print(f"   -> New best model saved with Val Acc: {val_acc:.4f}")

    return model, pd.DataFrame(train_history)

# CẢI TIẾN: Kiểm tra nếu có model đã huấn luyện thì tải, không thì huấn luyện mới
model_path = os.path.join(SAVE_DIR, 'dream_oe_best_model.pth')
dream_oe_model = get_model()
train_history = None

if os.path.exists(model_path):
    print(f"--- Tải mô hình DreamOE đã huấn luyện từ: {model_path} ---")
    dream_oe_model.load_state_dict(torch.load(model_path, map_location=DEVICE))
else:
    print(f"--- Bắt đầu huấn luyện mô hình DreamOE ({100} Epochs) ---") # CẢI TIẾN: Tăng epochs
    dream_oe_model, train_history = train_dream_oe_model(
        dream_oe_model, id_train_loader, id_test_loader, dream_ood_loader, epochs=100
    )

#==================================================================
# BƯỚC 5: ĐÁNH GIÁ CHUYÊN SÂU VÀ TRỰC QUAN HÓA
#==================================================================
print("\n" + " BƯỚC 5: ĐÁNH GIÁ CHUYÊN SÂU ".center(80, "="))

# Đánh giá mô hình DreamOE với các phương pháp
ood_methods = {'Energy': OODDetector.energy_score, 'MSP': OODDetector.max_softmax_probability}
results = []
for method_name, method_func in ood_methods.items():
    id_scores = method_func(dream_oe_model, id_test_loader)
    near_ood_scores = method_func(dream_oe_model, near_ood_loader)
    far_ood_scores = method_func(dream_oe_model, far_ood_loader)
    results.append({'Method': method_name, 'Dataset': 'Near-OOD', **evaluate_ood_performance(id_scores, near_ood_scores)})
    results.append({'Method': method_name, 'Dataset': 'Far-OOD', **evaluate_ood_performance(id_scores, far_ood_scores)})
results_df = pd.DataFrame(results)
print(results_df.to_string())

# CẢI TIẾN: Đồ thị mới, so sánh các phương pháp detect trên model DreamOE
def plot_dream_oe_performance(results_df, model):
    fig, axes = plt.subplots(1, 2, figsize=(18, 7))
    plt.suptitle('Hiệu suất Phát hiện OOD của Model DreamOE', fontsize=16)

    datasets = ['Near-OOD', 'Far-OOD']
    loaders = [near_ood_loader, far_ood_loader]

    for i, (dataset, loader) in enumerate(zip(datasets, loaders)):
        ax = axes[i]
        for method_name, method_func in ood_methods.items():
            id_scores = method_func(model, id_test_loader)
            ood_scores = method_func(model, loader)
            labels = np.concatenate([np.zeros_like(id_scores), np.ones_like(ood_scores)])
            scores = np.concatenate([id_scores, ood_scores])
            fpr, tpr, _ = roc_curve(labels, scores)
            auroc = auc(fpr, tpr)
            ax.plot(fpr, tpr, label=f'{method_name} (AUROC = {auroc:.4f})')

        ax.plot([0, 1], [0, 1], 'k--')
        ax.set_title(f'Đường cong ROC trên {dataset}')
        ax.set_xlabel('False Positive Rate'); ax.set_ylabel('True Positive Rate')
        ax.legend(); ax.grid(True)

    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.savefig(os.path.join(SAVE_DIR, 'dream_oe_performance_roc.png'))
    plt.show()
    wandb.log({"DreamOE ROC Performance": wandb.Image(plt)})

plot_dream_oe_performance(results_df, dream_oe_model)

#==================================================================
# BƯỚC 6: WEB APPLICATION TINH GỌN
#==================================================================
print("\n" + " BƯỚC 6: WEB APPLICATION TINH GỌN ".center(80, "="))
# (Class OODWebApp giữ nguyên, nhưng chỉ cần dùng với 1 model)
class OODWebApp:
    def __init__(self, model, model_name, id_classes_names, test_loader):
        self.model = model
        self.model_name = model_name
        self.id_classes_names = id_classes_names
        self.transform = transform_test
        self.threshold = self._calculate_threshold(test_loader)
    def _calculate_threshold(self, test_loader):
        print("Đang tính toán ngưỡng (threshold) cho OOD detection...")
        id_scores = OODDetector.energy_score(self.model, test_loader)
        threshold = np.percentile(id_scores, 95) # Ngưỡng tại percentile 95
        print(f"Ngưỡng Energy score được xác định là: {threshold:.4f}")
        return threshold
    def predict(self, image):
        pil_image = Image.fromarray(image).convert("RGB")
        image_tensor = self.transform(pil_image).unsqueeze(0).to(DEVICE)
        self.model.eval()
        with torch.no_grad():
            logits = self.model(image_tensor)
            ood_score = -torch.logsumexp(logits, dim=1).item()
            is_ood = ood_score > self.threshold

            if is_ood:
                return {"status": "Out-of-Distribution", "ood_score": ood_score, "is_ood": True}
            else:
                probs = F.softmax(logits, dim=1)[0]
                pred_idx = torch.argmax(probs).item()
                return {
                    "status": "In-Distribution", "ood_score": ood_score, "is_ood": False,
                    "predicted_class": self.id_classes_names[pred_idx], "confidence": probs[pred_idx].item()
                }

web_app = OODWebApp(dream_oe_model, "DreamOE", id_classes_names, id_test_loader)

def gradio_predict(image):
    result = web_app.predict(image)
    status = result["status"]
    score_info = f"Energy Score: {result['ood_score']:.4f} (Threshold: {web_app.threshold:.4f})"
    if result["is_ood"]:
        return f"🚨 **{status}**", "N/A", "N/A", score_info
    else:
        return f"✅ **{status}**", result["predicted_class"], f"{result['confidence']:.2%}", score_info

# CẢI TIẾN: Giao diện không cần chọn model
with gr.Blocks(title="🐾 OOD Animal Detection", theme=gr.themes.Soft()) as app:
    gr.Markdown("# 🐾 Out-of-Distribution Animal Detection\nUpload an image to detect if it's an animal or not, using the **DreamOE model**.")
    with gr.Row():
        with gr.Column():
            image_input = gr.Image(label="📸 Upload Image", type="numpy")
            predict_btn = gr.Button("🚀 Analyze Image", variant="primary")
        with gr.Column():
            status_output = gr.Markdown(label="📊 Detection Result")
            class_output = gr.Textbox(label="🏷️ Predicted Animal Class")
            confidence_output = gr.Textbox(label="🎯 Confidence")
            score_output = gr.Textbox(label="📈 OOD Score Details")
    predict_btn.click(fn=gradio_predict, inputs=image_input, outputs=[status_output, class_output, confidence_output, score_output])
    gr.Examples(
        [["cat.jpg"], ["truck.jpg"], ["qrcode.png"]],
        inputs=image_input,
        outputs=[status_output, class_output, confidence_output, score_output],
        fn=gradio_predict
    )

print("🚀 Launching Gradio Web Application...")
app.launch(share=True, debug=True)

print("\n🎉 PROJECT COMPLETED SUCCESSFULLY!")
```

```python
# -- coding: utf-8 --
"""
BÀI TOÁN NGOÀI PHÂN PHỐI ỨNG DỤNG TRONG NHẬN DIỆN MẪU
Phiên bản cải tiến toàn diện: Chống Overfitting, Pre-trained, Early Stopping, Mahalanobis
"""
#==================================================================
# BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG
#==================================================================
print(" BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG ".center(80, "="))
!pip install -q torchmetrics gradio wandb diffusers transformers accelerate opencv-python scikit-image
!git clone -q https://github.com/deeplearning-wisc/dream-ood.git

import os, sys, torch, torchvision
sys.path.append('./dream-ood')
import torch.nn as nn, torch.optim as optim, torch.nn.functional as F
import torchvision.transforms as T
from torch.utils.data import DataLoader, Subset, ConcatDataset, Dataset
from tqdm.notebook import tqdm
import numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.metrics import roc_auc_score, roc_curve, auc
from PIL import Image
import gradio as gr, pandas as pd, wandb
from datetime import datetime

wandb.init(project="ood-animal-detection-revamped", name=f"exp_{datetime.now().strftime('%Y%m%d_%H%M%S')}")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Sử dụng thiết bị: {DEVICE}")
from google.colab import drive
drive.mount('/content/drive')
SAVE_DIR = "/content/drive/MyDrive/OOD_Animal_Project_Revamped"
os.makedirs(SAVE_DIR, exist_ok=True)
print(f"Model và kết quả sẽ được lưu tại: {SAVE_DIR}")

#==================================================================
# BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU (Tối ưu hóa)
#==================================================================
print("\n" + " BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU ".center(80, "="))
class DreamOODGenerator:
    def __init__(self, model_name="stabilityai/stable-diffusion-2-1-base"):
        from diffusers import StableDiffusionPipeline
        self.pipe = StableDiffusionPipeline.from_pretrained(model_name, torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32)
        if torch.cuda.is_available(): self.pipe.to("cuda")
        self.ood_prompts = ["abstract geometric patterns", "random noise texture", "colorful abstract painting", "geometric fractal patterns", "digital glitch art"]
    def generate_ood_batch(self, batch_size=8, size=(32, 32)):
        images = []
        for _ in range(batch_size):
            prompt = np.random.choice(self.ood_prompts)
            random_elements = ["vibrant", "muted", "dark", "pastel"]
            prompt = f"{np.random.choice(random_elements)} {prompt}"
            with torch.no_grad():
                image = self.pipe(prompt, num_inference_steps=20, height=256, width=256, guidance_scale=7.5).images[0].resize(size)
            images.append(np.array(image))
        return np.array(images)
    def create_ood_dataset(self, num_samples=5000, save_path=None):
        print(f"Đang tạo {num_samples} mẫu OOD...")
        batch_size = 16
        num_batches = (num_samples + batch_size - 1) // batch_size
        all_images = np.concatenate([self.generate_ood_batch(min(batch_size, num_samples - i*batch_size)) for i in tqdm(range(num_batches), desc="Generating OOD")])
        if save_path: np.save(save_path, all_images); print(f"Đã lưu OOD dataset tại: {save_path}")
        return all_images

class DreamOODDataset(Dataset):
    def __init__(self, images, transform=None): self.images, self.transform = images, transform
    def __len__(self): return len(self.images)
    def __getitem__(self, idx):
        image = Image.fromarray(self.images[idx].astype(np.uint8))
        if self.transform: image = self.transform(image)
        return image, -1

normalize = T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
transform_train = T.Compose([T.RandomResizedCrop(32), T.RandomHorizontalFlip(), T.ColorJitter(brightness=0.2, contrast=0.2), T.ToTensor(), normalize])
transform_test = T.Compose([T.Resize(32), T.CenterCrop(32), T.ToTensor(), normalize])
transform_oe = T.Compose([T.Resize((32, 32)), T.RandomCrop(32, padding=4), T.RandomHorizontalFlip(), T.ToTensor(), T.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))])

dream_ood_path = os.path.join(SAVE_DIR, 'dream_ood_samples.npy')
if not os.path.exists(dream_ood_path):
    dream_generator = DreamOODGenerator(); dream_ood_images = dream_generator.create_ood_dataset(save_path=dream_ood_path)
else:
    print("Tải DreamOOD dataset đã có..."); dream_ood_images = np.load(dream_ood_path)
cifar10_classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']; cifar10_animal_indices = [2, 3, 4, 5, 6, 7]
cifar100_animal_indices = [1, 3, 4, 6, 7, 14, 15, 18, 19, 21, 26, 27, 30, 31, 32, 34, 36, 38, 42, 43, 44, 45, 46, 51, 56, 57, 65, 66, 67, 68, 70, 71, 72, 73, 74, 75, 76, 77, 86, 91, 93, 98, 99]; cifar100_classes = ['apple', 'aquarium_fish', 'baby', 'bear', 'beaver', 'bed', 'bee', 'beetle', 'bicycle', 'bottle', 'bowl', 'boy', 'bridge', 'bus', 'butterfly', 'camel', 'can', 'castle', 'caterpillar', 'cattle', 'chair', 'chimpanzee', 'clock', 'cloud', 'cockroach', 'couch', 'crab', 'crocodile', 'cup', 'dinosaur', 'dolphin', 'elephant', 'flatfish', 'forest', 'fox', 'girl', 'hamster', 'house', 'kangaroo', 'keyboard', 'lamp', 'lawn_mower', 'leopard', 'lion', 'lizard', 'lobster', 'man', 'maple_tree', 'motorcycle', 'mountain', 'mouse', 'mushroom', 'oak_tree', 'orange', 'orchid', 'otter', 'palm_tree', 'pear', 'pickup_truck', 'pine_tree', 'plain', 'plate', 'poppy', 'porcupine', 'possum', 'rabbit', 'raccoon', 'ray', 'road', 'rocket', 'rose', 'sea', 'seal', 'shark', 'shrew', 'skunk', 'skyscraper', 'snail', 'snake', 'spider', 'squirrel', 'streetcar', 'sunflower', 'sweet_pepper', 'table', 'tank', 'telephone', 'television', 'tiger', 'tractor', 'train', 'trout', 'tulip', 'turtle', 'wardrobe', 'whale', 'willow_tree', 'wolf', 'woman', 'worm']
id_classes_names = [cifar10_classes[i] for i in cifar10_animal_indices] + [cifar100_classes[i] for i in cifar100_animal_indices]
near_ood_classes_indices = [i for i in range(10) if i not in cifar10_animal_indices]
class RemappedSubset(Subset):
    def __init__(self, dataset, indices, label_map): super().__init__(dataset, indices); self.label_map = label_map
    def __getitem__(self, idx): image, label = self.dataset[self.indices[idx]]; return image, self.label_map[label]
cifar10_train=torchvision.datasets.CIFAR10('./data',train=True,download=True,transform=transform_train);cifar10_test=torchvision.datasets.CIFAR10('./data',train=False,download=True,transform=transform_test);cifar100_train=torchvision.datasets.CIFAR100('./data',train=True,download=True,transform=transform_train);cifar100_test=torchvision.datasets.CIFAR100('./data',train=False,download=True,transform=transform_test)
c10_label_map={orig:new for new,orig in enumerate(cifar10_animal_indices)};c100_label_map={orig:new+len(c10_label_map) for new,orig in enumerate(cifar100_animal_indices)}
id_train_dataset=ConcatDataset([RemappedSubset(cifar10_train,[i for i,(_,l)in enumerate(cifar10_train)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_train,[i for i,(_,l)in enumerate(cifar100_train)if l in cifar100_animal_indices],c100_label_map)])
id_test_dataset=ConcatDataset([RemappedSubset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_test,[i for i,(_,l)in enumerate(cifar100_test)if l in cifar100_animal_indices],c100_label_map)])
near_ood_test_dataset=Subset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in near_ood_classes_indices]);far_ood_test_dataset=torchvision.datasets.SVHN('./data',split='test',download=True,transform=transform_test);dream_ood_dataset=DreamOODDataset(dream_ood_images,transform=transform_oe)
batch_size=128;id_train_loader=DataLoader(id_train_dataset,batch_size,shuffle=True,num_workers=2,pin_memory=True);id_test_loader=DataLoader(id_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);near_ood_loader=DataLoader(near_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);far_ood_loader=DataLoader(far_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);dream_ood_loader=DataLoader(dream_ood_dataset,batch_size,shuffle=True,num_workers=2,pin_memory=True)

#==================================================================
# BƯỚC 3: CẢI TIẾN MODEL VÀ BỔ SUNG PHƯƠNG PHÁP OOD
#==================================================================
print("\n" + " BƯỚC 3: CẢI TIẾN MODEL VÀ BỔ SUNG PHƯƠNG PHÁP OOD ".center(80, "="))
class CustomResNet18(nn.Module):
    def __init__(self, num_classes, dropout_rate=0.5):
        super().__init__()
        self.backbone = torchvision.models.resnet18(weights=torchvision.models.ResNet18_Weights.DEFAULT)
        num_features = self.backbone.fc.in_features
        self.backbone.fc = nn.Identity()
        self.classifier = nn.Sequential(nn.Linear(num_features, 256), nn.ReLU(), nn.Dropout(p=dropout_rate), nn.Linear(256, num_classes))
    def forward(self, x, return_features=False):
        features = self.backbone(x); logits = self.classifier(features)
        if return_features: return logits, features
        return logits

def get_model(): return CustomResNet18(num_classes=len(id_classes_names)).to(DEVICE)
class OODDetector:
    @staticmethod
    def energy_score(model, data_loader):
        model.eval(); scores = []
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "Energy Score", leave=False):
                logits = model(images.to(DEVICE)); energy = -torch.logsumexp(logits, dim=1)
                scores.extend(energy.cpu().numpy())
        return np.array(scores)
    @staticmethod
    def max_softmax_probability(model, data_loader):
        model.eval(); scores = []
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "MSP", leave=False):
                probs = F.softmax(model(images.to(DEVICE)), dim=1); max_probs, _ = torch.max(probs, dim=1)
                scores.extend((-max_probs).cpu().numpy())
        return np.array(scores)
    class Mahalanobis:
        def fit(self, model, data_loader, num_classes):
            print("Fitting Mahalanobis detector..."); model.eval(); self.class_means = torch.zeros(num_classes, 256).to(DEVICE); class_counts = torch.zeros(num_classes).to(DEVICE); all_features = []
            with torch.no_grad():
                for images, labels in tqdm(data_loader, "Calculating means", leave=False):
                    images, labels = images.to(DEVICE), labels.to(DEVICE); _, features = model(images, return_features=True); all_features.append(features)
                    for i in range(num_classes):
                        class_features = features[labels == i]
                        if len(class_features) > 0: self.class_means[i] += class_features.sum(0); class_counts[i] += len(class_features)
            self.class_means /= class_counts.unsqueeze(1); all_features = torch.cat(all_features)
            all_labels = torch.cat([l for _, l in data_loader]).to(DEVICE); centered_features = all_features - self.class_means[all_labels]
            self.covariance = torch.cov(centered_features.T); self.precision = torch.linalg.inv(self.covariance); print("Mahalanobis fit complete.")
        def score(self, model, data_loader):
            print("Scoring with Mahalanobis..."); model.eval(); scores = []
            with torch.no_grad():
                for images, _ in tqdm(data_loader, "Mahalanobis Score", leave=False):
                    _, features = model(images.to(DEVICE), return_features=True); min_dists = []
                    for feature in features:
                        dists = [torch.sqrt((feature - mean) @ self.precision @ (feature - mean).T) for mean in self.class_means]
                        min_dists.append(torch.min(torch.stack(dists)))
                    scores.extend(torch.stack(min_dists).cpu().numpy())
            return np.array(scores)

def evaluate_ood_performance(id_scores, ood_scores):
    labels=np.concatenate([np.zeros_like(id_scores),np.ones_like(ood_scores)]);scores=np.concatenate([id_scores,ood_scores])
    auroc=roc_auc_score(labels,scores);fpr,tpr,_=roc_curve(labels,scores);fpr95=fpr[np.argmax(tpr>=0.95)]
    return {'auroc':auroc,'fpr95':fpr95}

#==================================================================
# BƯỚC 4: HUẤN LUYỆN NÂNG CAO VỚI EARLY STOPPING
#==================================================================
print("\n" + " BƯỚC 4: HUẤN LUYỆN NÂNG CAO ".center(80, "="))
class EarlyStopping:
    def __init__(self, patience=7, verbose=False, delta=0, path='checkpoint.pt'):
        self.patience, self.verbose, self.delta, self.path = patience, verbose, delta, path
        self.counter, self.best_score, self.early_stop = 0, None, False
        # [SỬA LỖI QUAN TRỌNG] Sử dụng np.inf thay vì np.Inf cho NumPy 2.0
        self.val_loss_min = np.inf
    def __call__(self, val_loss, model):
        score = -val_loss
        if self.best_score is None: self.best_score = score; self.save_checkpoint(val_loss, model)
        elif score < self.best_score + self.delta:
            self.counter += 1; print(f'EarlyStopping counter: {self.counter} out of {self.patience}')
            if self.counter >= self.patience: self.early_stop = True
        else: self.best_score = score; self.save_checkpoint(val_loss, model); self.counter = 0
    def save_checkpoint(self, val_loss, model):
        if self.verbose: print(f'Validation loss decreased ({self.val_loss_min:.6f} --> {val_loss:.6f}).  Saving model ...')
        torch.save(model.state_dict(), self.path); self.val_loss_min = val_loss

def train_dream_oe_model(model, train_loader, val_loader, ood_loader, epochs=100, ood_weight=0.5):
    optimizer = optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-4); scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs); criterion = nn.CrossEntropyLoss()
    early_stopping = EarlyStopping(patience=10, verbose=True, path=os.path.join(SAVE_DIR, 'dream_oe_best_model.pth')); train_history = []
    for epoch in range(epochs):
        model.train(); train_loss=0; train_correct=0; train_total=0; ood_iter = iter(ood_loader)
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}", leave=False)
        for inputs, targets in progress_bar:
            inputs,targets=inputs.to(DEVICE),targets.to(DEVICE); optimizer.zero_grad(); outputs = model(inputs); loss_id = criterion(outputs, targets)
            try: ood_inputs, _ = next(ood_iter)
            except StopIteration: ood_iter = iter(ood_loader); ood_inputs, _ = next(ood_iter)
            outputs_ood = model(ood_inputs.to(DEVICE)); loss_ood = -F.log_softmax(outputs_ood, dim=1).mean()
            total_loss = loss_id + ood_weight * loss_ood; total_loss.backward(); optimizer.step()
            train_loss+=total_loss.item(); _,predicted=outputs.max(1); train_total+=targets.size(0); train_correct+=predicted.eq(targets).sum().item()
        model.eval(); val_loss=0; val_correct=0; val_total=0
        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs,targets=inputs.to(DEVICE),targets.to(DEVICE); outputs=model(inputs); loss=criterion(outputs,targets)
                val_loss+=loss.item(); _,predicted=outputs.max(1); val_total+=targets.size(0); val_correct+=predicted.eq(targets).sum().item()
        avg_train_loss=train_loss/len(train_loader); avg_val_loss=val_loss/len(val_loader); train_acc,val_acc=train_correct/train_total,val_correct/val_total
        epoch_metrics={'epoch':epoch+1,'train_loss':avg_train_loss,'train_acc':train_acc,'val_loss':avg_val_loss,'val_acc':val_acc}
        train_history.append(epoch_metrics); wandb.log({f'dream_oe/{k}':v for k,v in epoch_metrics.items()})
        print(f"Epoch {epoch+1:03d}: Train Acc: {train_acc:.4f}, Val Acc: {val_acc:.4f} | Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}")
        scheduler.step(); early_stopping(avg_val_loss, model)
        if early_stopping.early_stop: print("Early stopping"); break
    model.load_state_dict(torch.load(os.path.join(SAVE_DIR, 'dream_oe_best_model.pth'))); return model, pd.DataFrame(train_history)

model_path = os.path.join(SAVE_DIR, 'dream_oe_best_model.pth')
dream_oe_model = get_model()
if os.path.exists(model_path):
    print(f"--- Tải mô hình đã huấn luyện từ: {model_path} ---"); dream_oe_model.load_state_dict(torch.load(model_path, map_location=DEVICE))
else:
    print(f"--- Bắt đầu huấn luyện mô hình DreamOE ---"); dream_oe_model, train_history = train_dream_oe_model(dream_oe_model, id_train_loader, id_test_loader, dream_ood_loader, epochs=100)

#==================================================================
# BƯỚC 5: ĐÁNH GIÁ CHUYÊN SÂU
#==================================================================
print("\n" + " BƯỚC 5: ĐÁNH GIÁ CHUYÊN SÂU ".center(80, "="))
mahalanobis_detector = OODDetector.Mahalanobis(); mahalanobis_detector.fit(dream_oe_model, id_train_loader, len(id_classes_names))
ood_methods = {'Energy': OODDetector.energy_score, 'MSP': OODDetector.max_softmax_probability, 'Mahalanobis': mahalanobis_detector.score}
results = []
for name, func in ood_methods.items():
    id_scores = func(dream_oe_model, id_test_loader); near_scores = func(dream_oe_model, near_ood_loader); far_scores = func(dream_oe_model, far_ood_loader)
    results.append({'Method':name,'Dataset':'Near-OOD',**evaluate_ood_performance(id_scores,near_scores)}); results.append({'Method':name,'Dataset':'Far-OOD',**evaluate_ood_performance(id_scores,far_scores)})
results_df = pd.DataFrame(results); print("\n" + " KẾT QUẢ ĐÁNH GIÁ ".center(80, "=")); print(results_df.to_string())
fig, axes = plt.subplots(1, 2, figsize=(18, 7)); plt.suptitle('Hiệu suất Phát hiện OOD của Model DreamOE', fontsize=16)
for i, dataset in enumerate(['Near-OOD', 'Far-OOD']):
    ax = axes[i]; data = results_df[results_df['Dataset'] == dataset]; sns.barplot(x='Method', y='auroc', data=data, ax=ax)
    ax.set_title(f'AUROC on {dataset}'); ax.set_ylabel('AUROC Score'); ax.set_ylim(0.5, 1.0)
    for p in ax.patches: ax.annotate(f'{p.get_height():.4f}',(p.get_x()+p.get_width()/2.,p.get_height()),ha='center',va='center',xytext=(0,5),textcoords='offset points')
plt.tight_layout(rect=[0,0,1,0.95]); plt.show(); wandb.log({"OOD Performance Comparison": wandb.Image(plt)})

#==================================================================
# BƯỚC 6: WEB APPLICATION
#==================================================================
print("\n" + " BƯỚC 6: WEB APPLICATION ".center(80, "="))
class OODWebApp:
    def __init__(self, model, id_classes_names, test_loader):
        self.model, self.id_classes_names, self.transform = model, id_classes_names, transform_test
        print("Đang tính toán ngưỡng (threshold) cho OOD detection..."); id_scores = OODDetector.energy_score(self.model, test_loader)
        self.threshold = np.percentile(id_scores, 95); print(f"Ngưỡng Energy score được xác định là: {self.threshold:.4f}")
    def predict(self, image):
        pil_image=Image.fromarray(image).convert("RGB"); image_tensor=self.transform(pil_image).unsqueeze(0).to(DEVICE); self.model.eval()
        with torch.no_grad():
            logits=self.model(image_tensor); ood_score=-torch.logsumexp(logits,dim=1).item(); is_ood=ood_score>self.threshold
            if is_ood: return {"status":"Out-of-Distribution","ood_score":ood_score,"is_ood":True}
            else:
                probs=F.softmax(logits,dim=1)[0]; pred_idx=torch.argmax(probs).item()
                return {"status":"In-Distribution","ood_score":ood_score,"is_ood":False,"predicted_class":self.id_classes_names[pred_idx],"confidence":probs[pred_idx].item()}

web_app = OODWebApp(dream_oe_model, id_classes_names, id_test_loader)
def gradio_predict(image):
    result=web_app.predict(image); score_info=f"Energy Score: {result['ood_score']:.4f} (Threshold: {web_app.threshold:.4f})"
    if result["is_ood"]: return f"🚨 **{result['status']}**","N/A","N/A",score_info
    else: return f"✅ **{result['status']}**",result["predicted_class"],f"{result['confidence']:.2%}",score_info

with gr.Blocks(title="🐾 OOD Animal Detection",theme=gr.themes.Soft()) as app:
    gr.Markdown("# 🐾 Out-of-Distribution Animal Detection\nUpload an image to detect if it's an animal or not, using the **Advanced DreamOE model**.")
    with gr.Row():
        with gr.Column():
            image_input=gr.Image(label="📸 Upload Image",type="numpy"); predict_btn=gr.Button("🚀 Analyze Image",variant="primary")
        with gr.Column():
            status_output=gr.Markdown(label="📊 Detection Result"); class_output=gr.Textbox(label="🏷️ Predicted Animal Class"); confidence_output=gr.Textbox(label="🎯 Confidence"); score_output=gr.Textbox(label="📈 OOD Score Details")
    predict_btn.click(fn=gradio_predict,inputs=image_input,outputs=[status_output,class_output,confidence_output,score_output])

print("🚀 Launching Gradio Web Application...")
app.launch(share=True, debug=True)
```

```python
# -- coding: utf-8 --
"""
BÀI TOÁN NGOÀI PHÂN PHỐI ỨNG DỤNG TRONG NHẬN DIỆN MẪU
Phiên bản cải tiến toàn diện: Chống Overfitting, Pre-trained, Early Stopping, Mahalanobis
"""
#==================================================================
# BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG
#==================================================================
print(" BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG ".center(80, "="))
!pip install -q torchmetrics gradio wandb diffusers transformers accelerate opencv-python scikit-image
!git clone -q https://github.com/deeplearning-wisc/dream-ood.git

import os, sys, torch, torchvision
sys.path.append('./dream-ood')
import torch.nn as nn, torch.optim as optim, torch.nn.functional as F
import torchvision.transforms as T
from torch.utils.data import DataLoader, Subset, ConcatDataset, Dataset
from tqdm.notebook import tqdm
import numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.metrics import roc_auc_score, roc_curve, auc
from PIL import Image
import gradio as gr, pandas as pd, wandb
from datetime import datetime

wandb.init(project="ood-animal-detection-revamped", name=f"exp_{datetime.now().strftime('%Y%m%d_%H%M%S')}")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Sử dụng thiết bị: {DEVICE}")
from google.colab import drive
drive.mount('/content/drive')
SAVE_DIR = "/content/drive/MyDrive/OOD_Animal_Project_Revamped"
os.makedirs(SAVE_DIR, exist_ok=True)
print(f"Model và kết quả sẽ được lưu tại: {SAVE_DIR}")

#==================================================================
# BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU (Tối ưu hóa)
#==================================================================
print("\n" + " BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU ".center(80, "="))
class DreamOODGenerator:
    def __init__(self, model_name="stabilityai/stable-diffusion-2-1-base"):
        from diffusers import StableDiffusionPipeline
        self.pipe = StableDiffusionPipeline.from_pretrained(model_name, torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32)
        if torch.cuda.is_available(): self.pipe.to("cuda")
        self.ood_prompts = ["abstract geometric patterns", "random noise texture", "colorful abstract painting", "geometric fractal patterns", "digital glitch art"]
    def generate_ood_batch(self, batch_size=8, size=(32, 32)):
        images = []
        for _ in range(batch_size):
            prompt = np.random.choice(self.ood_prompts)
            random_elements = ["vibrant", "muted", "dark", "pastel"]
            prompt = f"{np.random.choice(random_elements)} {prompt}"
            with torch.no_grad():
                image = self.pipe(prompt, num_inference_steps=20, height=256, width=256, guidance_scale=7.5).images[0].resize(size)
            images.append(np.array(image))
        return np.array(images)
    def create_ood_dataset(self, num_samples=5000, save_path=None):
        print(f"Đang tạo {num_samples} mẫu OOD...")
        batch_size = 16
        num_batches = (num_samples + batch_size - 1) // batch_size
        all_images = np.concatenate([self.generate_ood_batch(min(batch_size, num_samples - i*batch_size)) for i in tqdm(range(num_batches), desc="Generating OOD")])
        if save_path: np.save(save_path, all_images); print(f"Đã lưu OOD dataset tại: {save_path}")
        return all_images

class DreamOODDataset(Dataset):
    def __init__(self, images, transform=None): self.images, self.transform = images, transform
    def __len__(self): return len(self.images)
    def __getitem__(self, idx):
        image = Image.fromarray(self.images[idx].astype(np.uint8))
        if self.transform: image = self.transform(image)
        return image, -1

normalize = T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
transform_train = T.Compose([T.RandomResizedCrop(32), T.RandomHorizontalFlip(), T.ColorJitter(brightness=0.2, contrast=0.2), T.ToTensor(), normalize])
transform_test = T.Compose([T.Resize(32), T.CenterCrop(32), T.ToTensor(), normalize])
transform_oe = T.Compose([T.Resize((32, 32)), T.RandomCrop(32, padding=4), T.RandomHorizontalFlip(), T.ToTensor(), T.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))])

dream_ood_path = os.path.join(SAVE_DIR, 'dream_ood_samples.npy')
if not os.path.exists(dream_ood_path):
    dream_generator = DreamOODGenerator(); dream_ood_images = dream_generator.create_ood_dataset(save_path=dream_ood_path)
else:
    print("Tải DreamOOD dataset đã có..."); dream_ood_images = np.load(dream_ood_path)
cifar10_classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']; cifar10_animal_indices = [2, 3, 4, 5, 6, 7]
cifar100_animal_indices = [1, 3, 4, 6, 7, 14, 15, 18, 19, 21, 26, 27, 30, 31, 32, 34, 36, 38, 42, 43, 44, 45, 46, 51, 56, 57, 65, 66, 67, 68, 70, 71, 72, 73, 74, 75, 76, 77, 86, 91, 93, 98, 99]; cifar100_classes = ['apple', 'aquarium_fish', 'baby', 'bear', 'beaver', 'bed', 'bee', 'beetle', 'bicycle', 'bottle', 'bowl', 'boy', 'bridge', 'bus', 'butterfly', 'camel', 'can', 'castle', 'caterpillar', 'cattle', 'chair', 'chimpanzee', 'clock', 'cloud', 'cockroach', 'couch', 'crab', 'crocodile', 'cup', 'dinosaur', 'dolphin', 'elephant', 'flatfish', 'forest', 'fox', 'girl', 'hamster', 'house', 'kangaroo', 'keyboard', 'lamp', 'lawn_mower', 'leopard', 'lion', 'lizard', 'lobster', 'man', 'maple_tree', 'motorcycle', 'mountain', 'mouse', 'mushroom', 'oak_tree', 'orange', 'orchid', 'otter', 'palm_tree', 'pear', 'pickup_truck', 'pine_tree', 'plain', 'plate', 'poppy', 'porcupine', 'possum', 'rabbit', 'raccoon', 'ray', 'road', 'rocket', 'rose', 'sea', 'seal', 'shark', 'shrew', 'skunk', 'skyscraper', 'snail', 'snake', 'spider', 'squirrel', 'streetcar', 'sunflower', 'sweet_pepper', 'table', 'tank', 'telephone', 'television', 'tiger', 'tractor', 'train', 'trout', 'tulip', 'turtle', 'wardrobe', 'whale', 'willow_tree', 'wolf', 'woman', 'worm']
id_classes_names = [cifar10_classes[i] for i in cifar10_animal_indices] + [cifar100_classes[i] for i in cifar100_animal_indices]
near_ood_classes_indices = [i for i in range(10) if i not in cifar10_animal_indices]
class RemappedSubset(Subset):
    def __init__(self, dataset, indices, label_map): super().__init__(dataset, indices); self.label_map = label_map
    def __getitem__(self, idx): image, label = self.dataset[self.indices[idx]]; return image, self.label_map[label]
cifar10_train=torchvision.datasets.CIFAR10('./data',train=True,download=True,transform=transform_train);cifar10_test=torchvision.datasets.CIFAR10('./data',train=False,download=True,transform=transform_test);cifar100_train=torchvision.datasets.CIFAR100('./data',train=True,download=True,transform=transform_train);cifar100_test=torchvision.datasets.CIFAR100('./data',train=False,download=True,transform=transform_test)
c10_label_map={orig:new for new,orig in enumerate(cifar10_animal_indices)};c100_label_map={orig:new+len(c10_label_map) for new,orig in enumerate(cifar100_animal_indices)}
id_train_dataset=ConcatDataset([RemappedSubset(cifar10_train,[i for i,(_,l)in enumerate(cifar10_train)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_train,[i for i,(_,l)in enumerate(cifar100_train)if l in cifar100_animal_indices],c100_label_map)])
id_test_dataset=ConcatDataset([RemappedSubset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_test,[i for i,(_,l)in enumerate(cifar100_test)if l in cifar100_animal_indices],c100_label_map)])
near_ood_test_dataset=Subset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in near_ood_classes_indices]);far_ood_test_dataset=torchvision.datasets.SVHN('./data',split='test',download=True,transform=transform_test);dream_ood_dataset=DreamOODDataset(dream_ood_images,transform=transform_oe)
batch_size=128;id_train_loader=DataLoader(id_train_dataset,batch_size,shuffle=True,num_workers=2,pin_memory=True);id_test_loader=DataLoader(id_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);near_ood_loader=DataLoader(near_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);far_ood_loader=DataLoader(far_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);dream_ood_loader=DataLoader(dream_ood_dataset,batch_size,shuffle=True,num_workers=2,pin_memory=True)

#==================================================================
# BƯỚC 3: CẢI TIẾN MODEL VÀ BỔ SUNG PHƯƠNG PHÁP OOD
#==================================================================
print("\n" + " BƯỚC 3: CẢI TIẾN MODEL VÀ BỔ SUNG PHƯƠNG PHÁP OOD ".center(80, "="))
class CustomResNet18(nn.Module):
    def __init__(self, num_classes, dropout_rate=0.5):
        super().__init__()
        self.backbone = torchvision.models.resnet18(weights=torchvision.models.ResNet18_Weights.DEFAULT)
        num_features = self.backbone.fc.in_features
        self.backbone.fc = nn.Identity()
        self.classifier = nn.Sequential(nn.Linear(num_features, 256), nn.ReLU(), nn.Dropout(p=dropout_rate), nn.Linear(256, num_classes))
    def forward(self, x, return_features=False):
        features = self.backbone(x); logits = self.classifier(features)
        if return_features: return logits, features
        return logits

def get_model(): return CustomResNet18(num_classes=len(id_classes_names)).to(DEVICE)
class OODDetector:
    @staticmethod
    def energy_score(model, data_loader):
        model.eval(); scores = [];
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "Energy Score", leave=False):
                logits = model(images.to(DEVICE)); energy = -torch.logsumexp(logits, dim=1)
                scores.extend(energy.cpu().numpy())
        return np.array(scores)
    @staticmethod
    def max_softmax_probability(model, data_loader):
        model.eval(); scores = []
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "MSP", leave=False):
                probs = F.softmax(model(images.to(DEVICE)), dim=1); max_probs, _ = torch.max(probs, dim=1)
                scores.extend((-max_probs).cpu().numpy())
        return np.array(scores)

    class Mahalanobis:
        def fit(self, model, data_loader, num_classes):
            print("Fitting Mahalanobis detector..."); model.eval()
            try: feature_dim = model.classifier[0].in_features
            except (AttributeError, IndexError):
                dummy_input = torch.randn(2, 3, 32, 32).to(DEVICE); _, features = model(dummy_input, return_features=True); feature_dim = features.shape[1]
            print(f"Detected feature dimension: {feature_dim}")
            self.class_means = torch.zeros(num_classes, feature_dim).to(DEVICE); class_counts = torch.zeros(num_classes).to(DEVICE); all_features = []
            with torch.no_grad():
                for images, labels in tqdm(data_loader, "Calculating means", leave=False):
                    images, labels = images.to(DEVICE), labels.to(DEVICE); _, features = model(images, return_features=True); all_features.append(features)
                    for i in range(num_classes):
                        class_features = features[labels == i]
                        if len(class_features) > 0: self.class_means[i] += class_features.sum(0); class_counts[i] += len(class_features)
            class_counts[class_counts == 0] = 1; self.class_means /= class_counts.unsqueeze(1)
            all_features = torch.cat(all_features); all_labels = torch.cat([l for _, l in data_loader]).to(DEVICE)
            centered_features = all_features - self.class_means[all_labels]
            self.covariance = torch.cov(centered_features.T)

            # [SỬA LỖI QUAN TRỌNG] Thêm Regularization để tránh lỗi ma trận suy biến
            reg_covariance = self.covariance + torch.eye(feature_dim).to(DEVICE) * 1e-6
            self.precision = torch.linalg.inv(reg_covariance)

            print("Mahalanobis fit complete.")

        def score(self, model, data_loader):
            print("Scoring with Mahalanobis..."); model.eval(); scores = []
            with torch.no_grad():
                for images, _ in tqdm(data_loader, "Mahalanobis Score", leave=False):
                    _, features = model(images.to(DEVICE), return_features=True); min_dists = []
                    for feature in features:
                        dists = [torch.sqrt((feature - mean) @ self.precision @ (feature - mean).T) for mean in self.class_means]
                        min_dists.append(torch.min(torch.stack(dists)))
                    scores.extend(torch.stack(min_dists).cpu().numpy())
            return np.array(scores)

def evaluate_ood_performance(id_scores, ood_scores):
    labels=np.concatenate([np.zeros_like(id_scores),np.ones_like(ood_scores)]);scores=np.concatenate([id_scores,ood_scores])
    auroc=roc_auc_score(labels,scores);fpr,tpr,_=roc_curve(labels,scores);fpr95=fpr[np.argmax(tpr>=0.95)]
    return {'auroc':auroc,'fpr95':fpr95}

#==================================================================
# BƯỚC 4: HUẤN LUYỆN NÂNG CAO VỚI EARLY STOPPING
#==================================================================
print("\n" + " BƯỚC 4: HUẤN LUYỆN NÂNG CAO ".center(80, "="))
class EarlyStopping:
    def __init__(self, patience=7, verbose=False, delta=0, path='checkpoint.pt'):
        self.patience, self.verbose, self.delta, self.path = patience, verbose, delta, path; self.counter, self.best_score, self.early_stop = 0, None, False; self.val_loss_min = np.inf
    def __call__(self, val_loss, model):
        score = -val_loss
        if self.best_score is None: self.best_score = score; self.save_checkpoint(val_loss, model)
        elif score < self.best_score + self.delta:
            self.counter += 1; print(f'EarlyStopping counter: {self.counter} out of {self.patience}')
            if self.counter >= self.patience: self.early_stop = True
        else: self.best_score = score; self.save_checkpoint(val_loss, model); self.counter = 0
    def save_checkpoint(self, val_loss, model):
        if self.verbose: print(f'Validation loss decreased ({self.val_loss_min:.6f} --> {val_loss:.6f}).  Saving model ...')
        torch.save(model.state_dict(), self.path); self.val_loss_min = val_loss

def train_dream_oe_model(model, train_loader, val_loader, ood_loader, epochs=100, ood_weight=0.5):
    optimizer = optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-4); scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs); criterion = nn.CrossEntropyLoss()
    early_stopping = EarlyStopping(patience=10, verbose=True, path=os.path.join(SAVE_DIR, 'dream_oe_best_model.pth')); train_history = []
    for epoch in range(epochs):
        model.train(); train_loss=0; train_correct=0; train_total=0; ood_iter = iter(ood_loader)
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}", leave=False)
        for inputs, targets in progress_bar:
            inputs,targets=inputs.to(DEVICE),targets.to(DEVICE); optimizer.zero_grad(); outputs = model(inputs); loss_id = criterion(outputs, targets)
            try: ood_inputs, _ = next(ood_iter)
            except StopIteration: ood_iter = iter(ood_loader); ood_inputs, _ = next(ood_iter)
            outputs_ood = model(ood_inputs.to(DEVICE)); loss_ood = -F.log_softmax(outputs_ood, dim=1).mean()
            total_loss = loss_id + ood_weight * loss_ood; total_loss.backward(); optimizer.step()
            train_loss+=total_loss.item(); _,predicted=outputs.max(1); train_total+=targets.size(0); train_correct+=predicted.eq(targets).sum().item()
        model.eval(); val_loss=0; val_correct=0; val_total=0
        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs,targets=inputs.to(DEVICE),targets.to(DEVICE); outputs=model(inputs); loss=criterion(outputs,targets)
                val_loss+=loss.item(); _,predicted=outputs.max(1); val_total+=targets.size(0); val_correct+=predicted.eq(targets).sum().item()
        avg_train_loss=train_loss/len(train_loader); avg_val_loss=val_loss/len(val_loader); train_acc,val_acc=train_correct/train_total,val_correct/val_total
        epoch_metrics={'epoch':epoch+1,'train_loss':avg_train_loss,'train_acc':train_acc,'val_loss':avg_val_loss,'val_acc':val_acc}
        train_history.append(epoch_metrics); wandb.log({f'dream_oe/{k}':v for k,v in epoch_metrics.items()})
        print(f"Epoch {epoch+1:03d}: Train Acc: {train_acc:.4f}, Val Acc: {val_acc:.4f} | Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}")
        scheduler.step(); early_stopping(avg_val_loss, model)
        if early_stopping.early_stop: print("Early stopping"); break
    model.load_state_dict(torch.load(os.path.join(SAVE_DIR, 'dream_oe_best_model.pth'))); return model, pd.DataFrame(train_history)

model_path = os.path.join(SAVE_DIR, 'dream_oe_best_model.pth')
dream_oe_model = get_model()
if os.path.exists(model_path):
    print(f"--- Tải mô hình đã huấn luyện từ: {model_path} ---"); dream_oe_model.load_state_dict(torch.load(model_path, map_location=DEVICE))
else:
    print(f"--- Bắt đầu huấn luyện mô hình DreamOE ---"); dream_oe_model, train_history = train_dream_oe_model(dream_oe_model, id_train_loader, id_test_loader, dream_ood_loader, epochs=100)

#==================================================================
# BƯỚC 5: ĐÁNH GIÁ CHUYÊN SÂU
#==================================================================
print("\n" + " BƯỚC 5: ĐÁNH GIÁ CHUYÊN SÂU ".center(80, "="))
mahalanobis_detector = OODDetector.Mahalanobis(); mahalanobis_detector.fit(dream_oe_model, id_train_loader, len(id_classes_names))
ood_methods = {'Energy': OODDetector.energy_score, 'MSP': OODDetector.max_softmax_probability, 'Mahalanobis': mahalanobis_detector.score}
results = []
for name, func in ood_methods.items():
    id_scores = func(dream_oe_model, id_test_loader); near_scores = func(dream_oe_model, near_ood_loader); far_scores = func(dream_oe_model, far_ood_loader)
    results.append({'Method':name,'Dataset':'Near-OOD',**evaluate_ood_performance(id_scores,near_scores)}); results.append({'Method':name,'Dataset':'Far-OOD',**evaluate_ood_performance(id_scores,far_scores)})
results_df = pd.DataFrame(results); print("\n" + " KẾT QUẢ ĐÁNH GIÁ ".center(80, "=")); print(results_df.to_string())
fig, axes = plt.subplots(1, 2, figsize=(18, 7)); plt.suptitle('Hiệu suất Phát hiện OOD của Model DreamOE', fontsize=16)
for i, dataset in enumerate(['Near-OOD', 'Far-OOD']):
    ax = axes[i]; data = results_df[results_df['Dataset'] == dataset]; sns.barplot(x='Method', y='auroc', data=data, ax=ax)
    ax.set_title(f'AUROC on {dataset}'); ax.set_ylabel('AUROC Score'); ax.set_ylim(0.5, 1.0)
    for p in ax.patches: ax.annotate(f'{p.get_height():.4f}',(p.get_x()+p.get_width()/2.,p.get_height()),ha='center',va='center',xytext=(0,5),textcoords='offset points')
plt.tight_layout(rect=[0,0,1,0.95]); plt.show(); wandb.log({"OOD Performance Comparison": wandb.Image(plt)})

#==================================================================
# BƯỚC 6: WEB APPLICATION
#==================================================================
print("\n" + " BƯỚC 6: WEB APPLICATION ".center(80, "="))
class OODWebApp:
    def __init__(self, model, id_classes_names, test_loader):
        self.model, self.id_classes_names, self.transform = model, id_classes_names, transform_test
        print("Đang tính toán ngưỡng (threshold) cho OOD detection..."); id_scores = OODDetector.energy_score(self.model, test_loader)
        self.threshold = np.percentile(id_scores, 95); print(f"Ngưỡng Energy score được xác định là: {self.threshold:.4f}")
    def predict(self, image):
        pil_image=Image.fromarray(image).convert("RGB"); image_tensor=self.transform(pil_image).unsqueeze(0).to(DEVICE); self.model.eval()
        with torch.no_grad():
            logits=self.model(image_tensor); ood_score=-torch.logsumexp(logits,dim=1).item(); is_ood=ood_score>self.threshold
            if is_ood: return {"status":"Out-of-Distribution","ood_score":ood_score,"is_ood":True}
            else:
                probs=F.softmax(logits,dim=1)[0]; pred_idx=torch.argmax(probs).item()
                return {"status":"In-Distribution","ood_score":ood_score,"is_ood":False,"predicted_class":self.id_classes_names[pred_idx],"confidence":probs[pred_idx].item()}

web_app = OODWebApp(dream_oe_model, id_classes_names, id_test_loader)
def gradio_predict(image):
    result=web_app.predict(image); score_info=f"Energy Score: {result['ood_score']:.4f} (Threshold: {web_app.threshold:.4f})"
    if result["is_ood"]: return f"🚨 **{result['status']}**","N/A","N/A",score_info
    else: return f"✅ **{result['status']}**",result["predicted_class"],f"{result['confidence']:.2%}",score_info

with gr.Blocks(title="🐾 OOD Animal Detection",theme=gr.themes.Soft()) as app:
    gr.Markdown("# 🐾 Out-of-Distribution Animal Detection\nUpload an image to detect if it's an animal or not, using the **Advanced DreamOE model**.")
    with gr.Row():
        with gr.Column():
            image_input=gr.Image(label="📸 Upload Image",type="numpy"); predict_btn=gr.Button("🚀 Analyze Image",variant="primary")
        with gr.Column():
            status_output=gr.Markdown(label="📊 Detection Result"); class_output=gr.Textbox(label="🏷️ Predicted Animal Class"); confidence_output=gr.Textbox(label="🎯 Confidence"); score_output=gr.Textbox(label="📈 OOD Score Details")
    predict_btn.click(fn=gradio_predict,inputs=image_input,outputs=[status_output,class_output,confidence_output,score_output])

print("🚀 Launching Gradio Web Application...")
app.launch(share=True, debug=True)
```

```python
# -- coding: utf-8 --
"""
BÀI TOÁN NGOÀI PHÂN PHỐI ỨNG DỤNG TRONG NHẬN DIỆN MẪU
Phiên bản SOTA-inspired (đã sửa lỗi kiến trúc & NaN loss): Wide-ResNet + Supervised Contrastive Learning
"""
#==================================================================
# BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG
#==================================================================
print(" BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG ".center(80, "="))
!pip install -q torchmetrics gradio wandb diffusers transformers accelerate opencv-python scikit-image
!git clone -q https://github.com/deeplearning-wisc/dream-ood.git

import os, sys, torch, torchvision
sys.path.append('./dream-ood')
import torch.nn as nn, torch.optim as optim, torch.nn.functional as F
import torchvision.transforms as T
from torch.utils.data import DataLoader, Subset, ConcatDataset, Dataset
from tqdm.notebook import tqdm
import numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.metrics import roc_auc_score, roc_curve, auc
from PIL import Image
import gradio as gr, pandas as pd, wandb, math
from datetime import datetime

wandb.init(project="ood-animal-detection-sota-stable", name=f"exp_{datetime.now().strftime('%Y%m%d_%H%M%S')}")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Sử dụng thiết bị: {DEVICE}")
from google.colab import drive
drive.mount('/content/drive')
SAVE_DIR = "/content/drive/MyDrive/OOD_Animal_Project_Final"
os.makedirs(SAVE_DIR, exist_ok=True)
print(f"Model và kết quả sẽ được lưu tại: {SAVE_DIR}")

#==================================================================
# BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU
#==================================================================
print("\n" + " BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU ".center(80, "="))
class DreamOODGenerator:
    def __init__(self, model_name="stabilityai/stable-diffusion-2-1-base"):
        from diffusers import StableDiffusionPipeline
        self.pipe = StableDiffusionPipeline.from_pretrained(model_name, torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32)
        if torch.cuda.is_available(): self.pipe.to("cuda")
        self.ood_prompts = ["abstract geometric patterns", "random noise texture", "colorful abstract painting", "geometric fractal patterns", "digital glitch art"]
    def generate_ood_batch(self, batch_size=8, size=(32, 32)):
        images = []
        for _ in range(batch_size):
            prompt = np.random.choice(self.ood_prompts)
            random_elements = ["vibrant", "muted", "dark", "pastel"]
            prompt = f"{np.random.choice(random_elements)} {prompt}"
            with torch.no_grad():
                image = self.pipe(prompt, num_inference_steps=20, height=256, width=256, guidance_scale=7.5).images[0].resize(size)
            images.append(np.array(image))
        return np.array(images)
    def create_ood_dataset(self, num_samples=5000, save_path=None):
        print(f"Đang tạo {num_samples} mẫu OOD...")
        batch_size = 16
        num_batches = (num_samples + batch_size - 1) // batch_size
        all_images = np.concatenate([self.generate_ood_batch(min(batch_size, num_samples - i*batch_size)) for i in tqdm(range(num_batches), desc="Generating OOD")])
        if save_path: np.save(save_path, all_images); print(f"Đã lưu OOD dataset tại: {save_path}")
        return all_images
class DreamOODDataset(Dataset):
    def __init__(self, images, transform=None): self.images, self.transform = images, transform
    def __len__(self): return len(self.images)
    def __getitem__(self, idx):
        image = Image.fromarray(self.images[idx].astype(np.uint8))
        if self.transform: image = self.transform(image)
        return image, -1
normalize = T.Normalize(mean=[0.4914, 0.4822, 0.4465], std=[0.2023, 0.1994, 0.2010])
transform_train = T.Compose([T.RandomCrop(32, padding=4), T.RandomHorizontalFlip(), T.ToTensor(), normalize])
transform_test = T.Compose([T.ToTensor(), normalize])
dream_ood_path = os.path.join(SAVE_DIR, 'dream_ood_samples.npy')
if not os.path.exists(dream_ood_path):
    dream_generator = DreamOODGenerator(); dream_ood_images = dream_generator.create_ood_dataset(save_path=dream_ood_path)
else:
    print("Tải DreamOOD dataset đã có..."); dream_ood_images = np.load(dream_ood_path)
cifar10_classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']; cifar10_animal_indices = [2, 3, 4, 5, 6, 7]
cifar100_animal_indices = [1, 3, 4, 6, 7, 14, 15, 18, 19, 21, 26, 27, 30, 31, 32, 34, 36, 38, 42, 43, 44, 45, 46, 51, 56, 57, 65, 66, 67, 68, 70, 71, 72, 73, 74, 75, 76, 77, 86, 91, 93, 98, 99]; cifar100_classes = ['apple', 'aquarium_fish', 'baby', 'bear', 'beaver', 'bed', 'bee', 'beetle', 'bicycle', 'bottle', 'bowl', 'boy', 'bridge', 'bus', 'butterfly', 'camel', 'can', 'castle', 'caterpillar', 'cattle', 'chair', 'chimpanzee', 'clock', 'cloud', 'cockroach', 'couch', 'crab', 'crocodile', 'cup', 'dinosaur', 'dolphin', 'elephant', 'flatfish', 'forest', 'fox', 'girl', 'hamster', 'house', 'kangaroo', 'keyboard', 'lamp', 'lawn_mower', 'leopard', 'lion', 'lizard', 'lobster', 'man', 'maple_tree', 'motorcycle', 'mountain', 'mouse', 'mushroom', 'oak_tree', 'orange', 'orchid', 'otter', 'palm_tree', 'pear', 'pickup_truck', 'pine_tree', 'plain', 'plate', 'poppy', 'porcupine', 'possum', 'rabbit', 'raccoon', 'ray', 'road', 'rocket', 'rose', 'sea', 'seal', 'shark', 'shrew', 'skunk', 'skyscraper', 'snail', 'snake', 'spider', 'squirrel', 'streetcar', 'sunflower', 'sweet_pepper', 'table', 'tank', 'telephone', 'television', 'tiger', 'tractor', 'train', 'trout', 'tulip', 'turtle', 'wardrobe', 'whale', 'willow_tree', 'wolf', 'woman', 'worm']
id_classes_names = [cifar10_classes[i] for i in cifar10_animal_indices] + [cifar100_classes[i] for i in cifar100_animal_indices]
near_ood_classes_indices = [i for i in range(10) if i not in cifar10_animal_indices]
class RemappedSubset(Subset):
    def __init__(self, dataset, indices, label_map): super().__init__(dataset, indices); self.label_map = label_map
    def __getitem__(self, idx): image, label = self.dataset[self.indices[idx]]; return image, self.label_map[label]
cifar10_train=torchvision.datasets.CIFAR10('./data',train=True,download=True,transform=transform_train);cifar10_test=torchvision.datasets.CIFAR10('./data',train=False,download=True,transform=transform_test);cifar100_train=torchvision.datasets.CIFAR100('./data',train=True,download=True,transform=transform_train);cifar100_test=torchvision.datasets.CIFAR100('./data',train=False,download=True,transform=transform_test)
c10_label_map={orig:new for new,orig in enumerate(cifar10_animal_indices)};c100_label_map={orig:new+len(c10_label_map) for new,orig in enumerate(cifar100_animal_indices)}
id_train_dataset=ConcatDataset([RemappedSubset(cifar10_train,[i for i,(_,l)in enumerate(cifar10_train)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_train,[i for i,(_,l)in enumerate(cifar100_train)if l in cifar100_animal_indices],c100_label_map)])
id_test_dataset=ConcatDataset([RemappedSubset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_test,[i for i,(_,l)in enumerate(cifar100_test)if l in cifar100_animal_indices],c100_label_map)])
near_ood_test_dataset=Subset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in near_ood_classes_indices]);far_ood_test_dataset=torchvision.datasets.SVHN('./data',split='test',download=True,transform=transform_test);dream_ood_dataset=DreamOODDataset(dream_ood_images, T.Compose([T.ToTensor(), normalize]))
batch_size=128;id_train_loader=DataLoader(id_train_dataset,batch_size,shuffle=True,num_workers=2,pin_memory=True);id_test_loader=DataLoader(id_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);near_ood_loader=DataLoader(near_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);far_ood_loader=DataLoader(far_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);dream_ood_loader=DataLoader(dream_ood_dataset,batch_size,shuffle=True,num_workers=2,pin_memory=True)

#==================================================================
# BƯỚC 3: KIẾN TRÚC MODEL VÀ CÁC PHƯƠNG PHÁP OOD
#==================================================================
print("\n" + " BƯỚC 3: KIẾN TRÚC MODEL VÀ CÁC PHƯƠNG PHÁP OOD ".center(80, "="))
def conv3x3(in_planes, out_planes, stride=1): return nn.Conv2d(in_planes, out_planes, kernel_size=3, stride=stride, padding=1, bias=True)
class WideBasic(nn.Module):
    def __init__(self, in_planes, planes, dropout_rate, stride=1):
        super(WideBasic, self).__init__()
        self.bn1 = nn.BatchNorm2d(in_planes)
        # [SỬA LỖI KIẾN TRÚC] Truyền `stride` vào lớp conv đầu tiên
        self.conv1 = conv3x3(in_planes, planes, stride=stride)
        self.dropout = nn.Dropout(p=dropout_rate)
        self.bn2 = nn.BatchNorm2d(planes)
        self.conv2 = conv3x3(planes, planes)
        self.shortcut = nn.Sequential()
        if stride != 1 or in_planes != planes: self.shortcut = nn.Sequential(nn.Conv2d(in_planes, planes, kernel_size=1, stride=stride, bias=True),)
    def forward(self, x):
        out = self.dropout(self.conv1(F.relu(self.bn1(x)))); out = self.conv2(F.relu(self.bn2(out))); out += self.shortcut(x); return out
class WideResNet(nn.Module):
    def __init__(self, depth, widen_factor, dropout_rate, num_classes):
        super(WideResNet, self).__init__(); self.in_planes = 16; n = (depth - 4) / 6
        k = widen_factor; nStages = [16, 16*k, 32*k, 64*k]
        self.conv1 = conv3x3(3, nStages[0]); self.layer1 = self._make_layer(WideBasic, nStages[1], n, dropout_rate, stride=1)
        self.layer2 = self._make_layer(WideBasic, nStages[2], n, dropout_rate, stride=2); self.layer3 = self._make_layer(WideBasic, nStages[3], n, dropout_rate, stride=2)
        self.bn1 = nn.BatchNorm2d(nStages[3], momentum=0.9); self.linear = nn.Linear(nStages[3], num_classes)
        self.feature_dim = nStages[3]
    def _make_layer(self, block, planes, num_blocks, dropout_rate, stride):
        strides = [stride] + [1]*(int(num_blocks)-1); layers = []
        for stride in strides: layers.append(block(self.in_planes, planes, dropout_rate, stride)); self.in_planes = planes
        return nn.Sequential(*layers)
    def forward(self, x, return_features=False):
        out = self.conv1(x); out = self.layer1(out); out = self.layer2(out); out = self.layer3(out); out = F.relu(self.bn1(out))
        features = F.avg_pool2d(out, 8).view(out.size(0), -1); logits = self.linear(features)
        if return_features: return logits, features
        return logits
def get_model(): return WideResNet(depth=28, widen_factor=10, dropout_rate=0.3, num_classes=len(id_classes_names)).to(DEVICE)

class OODDetector: # (Giữ nguyên các phương pháp ổn định)
    @staticmethod
    def energy_score(model, data_loader):
        model.eval(); scores = [];
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "Energy Score", leave=False):
                logits = model(images.to(DEVICE)); energy = -torch.logsumexp(logits, dim=1)
                scores.extend(energy.cpu().numpy())
        return np.array(scores)
    @staticmethod
    def max_softmax_probability(model, data_loader):
        model.eval(); scores = []
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "MSP", leave=False):
                probs = F.softmax(model(images.to(DEVICE)), dim=1); max_probs, _ = torch.max(probs, dim=1)
                scores.extend((-max_probs).cpu().numpy())
        return np.array(scores)
    class Mahalanobis:
        def fit(self, model, data_loader, num_classes):
            print("Fitting Mahalanobis detector..."); model.eval()
            feature_dim = model.feature_dim
            print(f"Detected feature dimension: {feature_dim}")
            self.class_means = torch.zeros(num_classes, feature_dim).to(DEVICE); class_counts = torch.zeros(num_classes).to(DEVICE); all_features = []
            with torch.no_grad():
                for images, labels in tqdm(data_loader, "Calculating means", leave=False):
                    images, labels = images.to(DEVICE), labels.to(DEVICE); _, features = model(images, return_features=True); all_features.append(features)
                    for i in range(num_classes):
                        class_features = features[labels == i]
                        if len(class_features) > 0: self.class_means[i] += class_features.sum(0); class_counts[i] += len(class_features)
            class_counts[class_counts == 0] = 1; self.class_means /= class_counts.unsqueeze(1)
            all_features = torch.cat(all_features); all_labels = torch.cat([l for _, l in data_loader]).to(DEVICE)
            centered_features = all_features - self.class_means[all_labels]
            self.covariance = torch.cov(centered_features.T)
            reg_covariance = self.covariance + torch.eye(feature_dim).to(DEVICE) * 1e-6
            self.precision = torch.linalg.inv(reg_covariance)
            print("Mahalanobis fit complete.")
        def score(self, model, data_loader):
            print("Scoring with Mahalanobis..."); model.eval(); scores = []
            with torch.no_grad():
                for images, _ in tqdm(data_loader, "Mahalanobis Score", leave=False):
                    _, features = model(images.to(DEVICE), return_features=True); min_dists = []
                    for feature in features:
                        dists = [torch.sqrt((feature - mean) @ self.precision @ (feature - mean).T) for mean in self.class_means]
                        min_dists.append(torch.min(torch.stack(dists)))
                    scores.extend(torch.stack(min_dists).cpu().numpy())
            return np.array(scores)
def evaluate_ood_performance(id_scores, ood_scores):
    labels=np.concatenate([np.zeros_like(id_scores),np.ones_like(ood_scores)]);scores=np.concatenate([id_scores,ood_scores])
    auroc=roc_auc_score(labels,scores);fpr,tpr,_=roc_curve(labels,scores);fpr95=fpr[np.argmax(tpr>=0.95)]
    return {'auroc':auroc,'fpr95':fpr95}

#==================================================================
# BƯỚC 4: HUẤN LUYỆN NÂNG CAO VỚI SUPERVISED CONTRASTIVE LOSS
#==================================================================
print("\n" + " BƯỚC 4: HUẤN LUYỆN NÂNG CAO VỚI SUPERVISED CONTRASTIVE LOSS ".center(80, "="))
class SupConLoss(nn.Module):
    def __init__(self, temperature=0.07):
        super(SupConLoss, self).__init__(); self.temperature = temperature
    def forward(self, features, labels):
        features = F.normalize(features, dim=1); batch_size = features.shape[0]
        mask = torch.eq(labels.unsqueeze(1), labels.unsqueeze(0).T).float().to(DEVICE)
        anchor_dot_contrast = torch.div(torch.matmul(features, features.T), self.temperature)
        logits_mask = torch.scatter(torch.ones_like(mask), 1, torch.arange(batch_size).view(-1, 1).to(DEVICE), 0)
        mask = mask * logits_mask
        log_prob = anchor_dot_contrast - torch.log((logits_mask * torch.exp(anchor_dot_contrast)).sum(1, keepdim=True))
        # [SỬA LỖI NaN] Thêm epsilon và lọc các giá trị hữu hạn
        mean_log_prob_pos = (mask * log_prob).sum(1) / (mask.sum(1) + 1e-6)
        finite_vals = mean_log_prob_pos[mean_log_prob_pos.isfinite()]
        if len(finite_vals) == 0: return torch.tensor(0.0).to(DEVICE)
        loss = -finite_vals.mean(); return loss

class EarlyStopping:
    def __init__(self, patience=7, verbose=False, delta=0, path='checkpoint.pt'):
        self.patience, self.verbose, self.delta, self.path = patience, verbose, delta, path; self.counter, self.best_score, self.early_stop = 0, None, False; self.val_loss_min = np.inf
    def __call__(self, val_loss, model):
        score = -val_loss
        if self.best_score is None: self.best_score = score; self.save_checkpoint(val_loss, model)
        elif score < self.best_score + self.delta:
            self.counter += 1; print(f'EarlyStopping counter: {self.counter} out of {self.patience}')
            if self.counter >= self.patience: self.early_stop = True
        else: self.best_score = score; self.save_checkpoint(val_loss, model); self.counter = 0
    def save_checkpoint(self, val_loss, model):
        if self.verbose: print(f'Validation loss decreased ({self.val_loss_min:.6f} --> {val_loss:.6f}).  Saving model ...')
        torch.save(model.state_dict(), self.path); self.val_loss_min = val_loss

def train_contrastive_model(model, train_loader, val_loader, ood_loader, epochs=200):
    optimizer = optim.SGD(model.parameters(), lr=0.1, momentum=0.9, weight_decay=5e-4)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)
    criterion_ce = nn.CrossEntropyLoss()
    criterion_supcon = SupConLoss(temperature=0.1)
    early_stopping = EarlyStopping(patience=20, verbose=True, path=os.path.join(SAVE_DIR, 'contrastive_best_model.pth'))
    contrastive_weight = 0.5

    for epoch in range(epochs):
        model.train(); train_loss=0; train_correct=0; train_total=0; ood_iter = iter(ood_loader)
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}", leave=False)
        for id_inputs, id_targets in progress_bar:
            id_inputs, id_targets = id_inputs.to(DEVICE), id_targets.to(DEVICE)
            try: ood_inputs, _ = next(ood_iter)
            except StopIteration: ood_iter = iter(ood_loader); ood_inputs, _ = next(ood_iter)
            ood_inputs = ood_inputs.to(DEVICE)

            inputs = torch.cat([id_inputs, ood_inputs]); optimizer.zero_grad()
            logits, features = model(inputs, return_features=True)
            id_logits = logits[:len(id_inputs)]

            loss_ce = criterion_ce(id_logits, id_targets)
            contrastive_labels = torch.cat([id_targets, torch.full((len(ood_inputs),), len(id_classes_names)).to(DEVICE)])
            loss_supcon = criterion_supcon(features, contrastive_labels)

            total_loss = loss_ce + contrastive_weight * loss_supcon
            # [SỬA LỖI NaN] Chốt an toàn trước khi backward
            if not torch.isnan(total_loss):
                total_loss.backward(); optimizer.step(); train_loss+=total_loss.item()

            _,predicted=id_logits.max(1); train_total+=id_targets.size(0); train_correct+=predicted.eq(id_targets).sum().item()

        model.eval(); val_loss=0; val_correct=0; val_total=0
        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs,targets=inputs.to(DEVICE),targets.to(DEVICE); outputs=model(inputs); loss=criterion_ce(outputs,targets)
                val_loss+=loss.item(); _,predicted=outputs.max(1); val_total+=targets.size(0); val_correct+=predicted.eq(targets).sum().item()

        avg_train_loss=train_loss/len(train_loader); avg_val_loss=val_loss/len(val_loader); train_acc,val_acc=train_correct/train_total,val_correct/val_total
        epoch_metrics={'epoch':epoch+1,'train_loss':avg_train_loss,'train_acc':train_acc,'val_loss':avg_val_loss,'val_acc':val_acc}
        wandb.log({f'contrastive/{k}':v for k,v in epoch_metrics.items()}); print(f"Epoch {epoch+1:03d}: Train Acc: {train_acc:.4f}, Val Acc: {val_acc:.4f} | Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}")
        scheduler.step(); early_stopping(avg_val_loss, model)
        if early_stopping.early_stop: print("Early stopping"); break

    model.load_state_dict(torch.load(os.path.join(SAVE_DIR, 'contrastive_best_model.pth'))); return model

model_path = os.path.join(SAVE_DIR, 'contrastive_best_model.pth')
model = get_model()
if os.path.exists(model_path):
    print(f"--- Tải mô hình đã huấn luyện từ: {model_path} ---"); model.load_state_dict(torch.load(model_path, map_location=DEVICE))
else:
    print(f"--- Bắt đầu huấn luyện mô hình Contrastive ---"); model = train_contrastive_model(model, id_train_loader, id_test_loader, dream_ood_loader, epochs=200)

#==================================================================
# BƯỚC 5: ĐÁNH GIÁ CHUYÊN SÂU
#==================================================================
print("\n" + " BƯỚC 5: ĐÁNH GIÁ CHUYÊN SÂU ".center(80, "="))
mahalanobis_detector = OODDetector.Mahalanobis(); mahalanobis_detector.fit(model, id_train_loader, len(id_classes_names))
ood_methods = {'Energy': OODDetector.energy_score, 'MSP': OODDetector.max_softmax_probability, 'Mahalanobis': mahalanobis_detector.score}
results = []
for name, func in ood_methods.items():
    id_scores = func(model, id_test_loader); near_scores = func(model, near_ood_loader); far_scores = func(model, far_ood_loader)
    results.append({'Method':name,'Dataset':'Near-OOD',**evaluate_ood_performance(id_scores,near_scores)}); results.append({'Method':name,'Dataset':'Far-OOD',**evaluate_ood_performance(id_scores,far_scores)})
results_df = pd.DataFrame(results); print("\n" + " KẾT QUẢ ĐÁNH GIÁ ".center(80, "=")); print(results_df.to_string())
fig, axes = plt.subplots(1, 2, figsize=(18, 7)); plt.suptitle('Hiệu suất Phát hiện OOD của Model Contrastive', fontsize=16)
for i, dataset in enumerate(['Near-OOD', 'Far-OOD']):
    ax = axes[i]; data = results_df[results_df['Dataset'] == dataset]; sns.barplot(x='Method', y='auroc', data=data, ax=ax)
    ax.set_title(f'AUROC on {dataset}'); ax.set_ylabel('AUROC Score'); ax.set_ylim(0.7, 1.0)
    for p in ax.patches: ax.annotate(f'{p.get_height():.4f}',(p.get_x()+p.get_width()/2.,p.get_height()),ha='center',va='center',xytext=(0,5),textcoords='offset points')
plt.tight_layout(rect=[0,0,1,0.95]); plt.show(); wandb.log({"OOD Performance Comparison": wandb.Image(plt)})

#==================================================================
# BƯỚC 6: WEB APPLICATION
#==================================================================
print("\n" + " BƯỚC 6: WEB APPLICATION ".center(80, "="))
class OODWebApp:
    def __init__(self, model, id_classes_names, test_loader, results_df):
        self.model, self.id_classes_names, self.transform = model, id_classes_names, transform_test
        best_method_name = results_df.loc[results_df['auroc'].idxmax()]['Method']
        if best_method_name == 'Mahalanobis': best_method_name = 'Energy' # Fallback
        self.best_method = getattr(OODDetector, {'Energy':'energy_score', 'MSP':'max_softmax_probability'}[best_method_name])
        self.method_name = best_method_name
        print(f"Phương pháp OOD tốt nhất được chọn cho web app: {self.method_name}")
        print("Đang tính toán ngưỡng (threshold)...");
        id_scores = self.best_method(self.model, test_loader)
        self.threshold = np.percentile(id_scores, 95)
        print(f"Ngưỡng cho {self.method_name} được xác định là: {self.threshold:.4f}")
    def predict(self, image):
        pil_image=Image.fromarray(image).convert("RGB"); image_tensor=self.transform(pil_image).unsqueeze(0).to(DEVICE); self.model.eval()
        with torch.no_grad():
            logits = self.model(image_tensor)
            if self.method_name == 'Energy': ood_score = -torch.logsumexp(logits, dim=1).item()
            else: ood_score = -torch.max(F.softmax(logits, dim=1)).item()
            is_ood=ood_score > self.threshold
            if is_ood: return {"status":"Out-of-Distribution", "ood_score":ood_score, "is_ood":True}
            else:
                probs=F.softmax(logits,dim=1)[0]; pred_idx=torch.argmax(probs).item()
                return {"status":"In-Distribution", "ood_score":ood_score, "is_ood":False, "predicted_class":self.id_classes_names[pred_idx], "confidence":probs[pred_idx].item()}

web_app = OODWebApp(model, id_classes_names, id_test_loader, results_df)
def gradio_predict(image):
    result=web_app.predict(image);
    score_info=f"{web_app.method_name} Score: {result['ood_score']:.4f} (Threshold: {web_app.threshold:.4f})"
    if result["is_ood"]: return f"🚨 **{result['status']}**","N/A","N/A",score_info
    else: return f"✅ **{result['status']}**",result["predicted_class"],f"{result['confidence']:.2%}",score_info

with gr.Blocks(title="🐾 OOD Animal Detection",theme=gr.themes.Soft()) as app:
    gr.Markdown("# 🐾 Out-of-Distribution Animal Detection\nUpload an image to detect if it's an animal or not, using an **Advanced Contrastive Learning model**.")
    with gr.Row():
        with gr.Column():
            image_input=gr.Image(label="📸 Upload Image",type="numpy"); predict_btn=gr.Button("🚀 Analyze Image",variant="primary")
        with gr.Column():
            status_output=gr.Markdown(label="📊 Detection Result"); class_output=gr.Textbox(label="🏷️ Predicted Animal Class"); confidence_output=gr.Textbox(label="🎯 Confidence"); score_output=gr.Textbox(label="📈 OOD Score Details")
    predict_btn.click(fn=gradio_predict,inputs=image_input,outputs=[status_output,class_output,confidence_output,score_output])

print("🚀 Launching Gradio Web Application...")
app.launch(share=True, debug=True)
```

```python
"""
BÀI TOÁN NGOÀI PHÂN PHỐI ỨNG DỤNG TRONG NHẬN DIỆN MẪU
Phiên bản cuối cùng (đã sửa lỗi DataLoader): Wide-ResNet + ReAct (SOTA)
"""
#==================================================================
# BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG
#==================================================================
print(" BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG ".center(80, "="))
!pip install -q torchmetrics gradio wandb diffusers transformers accelerate opencv-python scikit-image
!git clone -q https://github.com/deeplearning-wisc/dream-ood.git

import os, sys, torch, torchvision
sys.path.append('./dream-ood')
import torch.nn as nn, torch.optim as optim, torch.nn.functional as F
import torchvision.transforms as T
from torch.utils.data import DataLoader, Subset, ConcatDataset, Dataset
from tqdm.notebook import tqdm
import numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.metrics import roc_auc_score, roc_curve, auc
from PIL import Image
import gradio as gr, pandas as pd, wandb
from datetime import datetime

wandb.init(project="ood-animal-detection-react-final", name=f"exp_{datetime.now().strftime('%Y%m%d_%H%M%S')}")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Sử dụng thiết bị: {DEVICE}")
from google.colab import drive
drive.mount('/content/drive')
SAVE_DIR = "/content/drive/MyDrive/OOD_Animal_Project_ReAct"
os.makedirs(SAVE_DIR, exist_ok=True)
print(f"Model và kết quả sẽ được lưu tại: {SAVE_DIR}")

#==================================================================
# BƯỚC 1: CHUẨN BỊ DỮ LIỆU
#==================================================================
print("\n" + " BƯỚC 1: CHUẨN BỊ DỮ LIỆU ".center(80, "="))
normalize = T.Normalize(mean=[0.4914, 0.4822, 0.4465], std=[0.2023, 0.1994, 0.2010])
transform_train = T.Compose([T.RandomCrop(32, padding=4), T.RandomHorizontalFlip(), T.ToTensor(), normalize])
transform_test = T.Compose([T.ToTensor(), normalize])
cifar10_classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']; cifar10_animal_indices = [2, 3, 4, 5, 6, 7]
cifar100_animal_indices = [1, 3, 4, 6, 7, 14, 15, 18, 19, 21, 26, 27, 30, 31, 32, 34, 36, 38, 42, 43, 44, 45, 46, 51, 56, 57, 65, 66, 67, 68, 70, 71, 72, 73, 74, 75, 76, 77, 86, 91, 93, 98, 99]; cifar100_classes = ['apple', 'aquarium_fish', 'baby', 'bear', 'beaver', 'bed', 'bee', 'beetle', 'bicycle', 'bottle', 'bowl', 'boy', 'bridge', 'bus', 'butterfly', 'camel', 'can', 'castle', 'caterpillar', 'cattle', 'chair', 'chimpanzee', 'clock', 'cloud', 'cockroach', 'couch', 'crab', 'crocodile', 'cup', 'dinosaur', 'dolphin', 'elephant', 'flatfish', 'forest', 'fox', 'girl', 'hamster', 'house', 'kangaroo', 'keyboard', 'lamp', 'lawn_mower', 'leopard', 'lion', 'lizard', 'lobster', 'man', 'maple_tree', 'motorcycle', 'mountain', 'mouse', 'mushroom', 'oak_tree', 'orange', 'orchid', 'otter', 'palm_tree', 'pear', 'pickup_truck', 'pine_tree', 'plain', 'plate', 'poppy', 'porcupine', 'possum', 'rabbit', 'raccoon', 'ray', 'road', 'rocket', 'rose', 'sea', 'seal', 'shark', 'shrew', 'skunk', 'skyscraper', 'snail', 'snake', 'spider', 'squirrel', 'streetcar', 'sunflower', 'sweet_pepper', 'table', 'tank', 'telephone', 'television', 'tiger', 'tractor', 'train', 'trout', 'tulip', 'turtle', 'wardrobe', 'whale', 'willow_tree', 'wolf', 'woman', 'worm']
id_classes_names = [cifar10_classes[i] for i in cifar10_animal_indices] + [cifar100_classes[i] for i in cifar100_animal_indices]
near_ood_classes_indices = [i for i in range(10) if i not in cifar10_animal_indices]
class RemappedSubset(Subset):
    def __init__(self, dataset, indices, label_map): super().__init__(dataset, indices); self.label_map = label_map
    def __getitem__(self, idx): image, label = self.dataset[self.indices[idx]]; return image, self.label_map[label]
cifar10_train=torchvision.datasets.CIFAR10('./data',train=True,download=True,transform=transform_train);cifar10_test=torchvision.datasets.CIFAR10('./data',train=False,download=True,transform=transform_test);cifar100_train=torchvision.datasets.CIFAR100('./data',train=True,download=True,transform=transform_train);cifar100_test=torchvision.datasets.CIFAR100('./data',train=False,download=True,transform=transform_test)
c10_label_map={orig:new for new,orig in enumerate(cifar10_animal_indices)};c100_label_map={orig:new+len(c10_label_map) for new,orig in enumerate(cifar100_animal_indices)}
id_train_dataset=ConcatDataset([RemappedSubset(cifar10_train,[i for i,(_,l)in enumerate(cifar10_train)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_train,[i for i,(_,l)in enumerate(cifar100_train)if l in cifar100_animal_indices],c100_label_map)])
id_test_dataset=ConcatDataset([RemappedSubset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_test,[i for i,(_,l)in enumerate(cifar100_test)if l in cifar100_animal_indices],c100_label_map)])
near_ood_test_dataset=Subset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in near_ood_classes_indices]);far_ood_test_dataset=torchvision.datasets.SVHN('./data',split='test',download=True,transform=transform_test)
batch_size=128;id_train_loader=DataLoader(id_train_dataset,batch_size,shuffle=True,num_workers=2,pin_memory=True);id_test_loader=DataLoader(id_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);near_ood_loader=DataLoader(near_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);far_ood_loader=DataLoader(far_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True)

#==================================================================
# BƯỚC 2: KIẾN TRÚC MODEL VỚI REACT
#==================================================================
print("\n" + " BƯỚC 2: KIẾN TRÚC MODEL VỚI REACT ".center(80, "="))
def conv3x3(in_planes, out_planes, stride=1): return nn.Conv2d(in_planes, out_planes, kernel_size=3, stride=stride, padding=1, bias=True)
class WideBasic(nn.Module):
    def __init__(self, in_planes, planes, dropout_rate, stride=1):
        super(WideBasic, self).__init__(); self.bn1 = nn.BatchNorm2d(in_planes); self.conv1 = conv3x3(in_planes, planes, stride); self.dropout = nn.Dropout(p=dropout_rate); self.bn2 = nn.BatchNorm2d(planes); self.conv2 = conv3x3(planes, planes); self.shortcut = nn.Sequential()
        if stride != 1 or in_planes != planes: self.shortcut = nn.Sequential(nn.Conv2d(in_planes, planes, kernel_size=1, stride=stride, bias=True),)
    def forward(self, x): out = self.dropout(self.conv1(F.relu(self.bn1(x)))); out = self.conv2(F.relu(self.bn2(out))); out += self.shortcut(x); return out
class WideResNet(nn.Module):
    def __init__(self, depth, widen_factor, dropout_rate, num_classes):
        super(WideResNet, self).__init__(); self.in_planes = 16; n = (depth - 4) / 6; k = widen_factor; nStages = [16, 16*k, 32*k, 64*k]
        self.conv1 = conv3x3(3, nStages[0]); self.layer1 = self._make_layer(WideBasic, nStages[1], n, dropout_rate, stride=1); self.layer2 = self._make_layer(WideBasic, nStages[2], n, dropout_rate, stride=2); self.layer3 = self._make_layer(WideBasic, nStages[3], n, dropout_rate, stride=2)
        self.bn1 = nn.BatchNorm2d(nStages[3], momentum=0.9); self.linear = nn.Linear(nStages[3], num_classes); self.feature_dim = nStages[3]
    def _make_layer(self, block, planes, num_blocks, dropout_rate, stride):
        strides = [stride] + [1]*(int(num_blocks)-1); layers = []
        for stride in strides: layers.append(block(self.in_planes, planes, dropout_rate, stride)); self.in_planes = planes
        return nn.Sequential(*layers)
    def get_features(self, x):
        out = self.conv1(x); out = self.layer1(out); out = self.layer2(out); out = self.layer3(out); out = F.relu(self.bn1(out)); return F.avg_pool2d(out, 8).view(out.size(0), -1)
    def forward(self, x): features = self.get_features(x); return self.linear(features)
class ReActModel(nn.Module):
    def __init__(self, num_classes, percentile=90):
        super().__init__(); self.backbone = WideResNet(depth=28, widen_factor=10, dropout_rate=0.3, num_classes=num_classes)
        self.percentile = percentile; self.threshold = None
    def forward(self, x, apply_react=False):
        features = self.backbone.get_features(x)
        if apply_react and self.threshold is not None: features = torch.clip(features, max=self.threshold)
        return self.backbone.linear(features)
    def calibrate(self, data_loader):
        print(f"Calibrating ReAct threshold at {self.percentile}th percentile..."); self.backbone.eval(); all_features = []
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "Calibration", leave=False):
                features = self.backbone.get_features(images.to(DEVICE)); all_features.append(features)
        all_features = torch.cat(all_features); self.threshold = torch.quantile(all_features, self.percentile / 100.0)
        print(f"ReAct threshold set to: {self.threshold.item():.4f}")
    def ood_score(self, x):
        with torch.no_grad():
            logits_normal = self.forward(x, apply_react=False); logits_react = self.forward(x, apply_react=True)
            energy_normal = -torch.logsumexp(logits_normal, dim=1); energy_react = -torch.logsumexp(logits_react, dim=1)
            return (energy_react - energy_normal).cpu().numpy()
def get_model(): return ReActModel(num_classes=len(id_classes_names)).to(DEVICE)

#==================================================================
# BƯỚC 3: QUY TRÌNH HUẤN LUYỆN
#==================================================================
print("\n" + " BƯỚC 3: QUY TRÌNH HUẤN LUYỆN ".center(80, "="))
class EarlyStopping:
    def __init__(self, patience=7, verbose=False, path='checkpoint.pt'):
        self.patience, self.verbose, self.path = patience, verbose, path; self.counter, self.best_score, self.early_stop = 0, None, False; self.val_loss_min = np.inf
    def __call__(self, val_loss, model):
        score = -val_loss
        if self.best_score is None: self.best_score = score; self.save_checkpoint(val_loss, model)
        elif score < self.best_score:
            self.counter += 1; print(f'EarlyStopping counter: {self.counter} out of {self.patience}')
            if self.counter >= self.patience: self.early_stop = True
        else: self.best_score = score; self.save_checkpoint(val_loss, model); self.counter = 0
    def save_checkpoint(self, val_loss, model):
        if self.verbose: print(f'Validation loss decreased ({self.val_loss_min:.6f} --> {val_loss:.6f}).  Saving model ...')
        torch.save(model.state_dict(), self.path); self.val_loss_min = val_loss

def train_model(model, train_loader, val_loader, epochs=200):
    optimizer = optim.SGD(model.parameters(), lr=0.1, momentum=0.9, weight_decay=5e-4)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)
    criterion = nn.CrossEntropyLoss()
    early_stopping = EarlyStopping(patience=20, verbose=True, path=os.path.join(SAVE_DIR, 'react_wrn_best.pth'))
    for epoch in range(epochs):
        model.train(); train_loss=0; train_correct=0; train_total=0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}", leave=False)
        for inputs, targets in progress_bar:
            inputs,targets=inputs.to(DEVICE),targets.to(DEVICE); optimizer.zero_grad(); outputs = model(inputs)
            loss = criterion(outputs, targets); loss.backward(); optimizer.step()
            train_loss+=loss.item(); _,predicted=outputs.max(1); train_total+=targets.size(0); train_correct+=predicted.eq(targets).sum().item()
        model.eval(); val_loss=0; val_correct=0; val_total=0
        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs,targets=inputs.to(DEVICE),targets.to(DEVICE); outputs=model(inputs); loss=criterion(outputs,targets)
                val_loss+=loss.item(); _,predicted=outputs.max(1); val_total+=targets.size(0); val_correct+=predicted.eq(targets).sum().item()
        avg_train_loss=train_loss/len(train_loader); avg_val_loss=val_loss/len(val_loader); train_acc,val_acc=train_correct/train_total,val_correct/val_total
        epoch_metrics={'epoch':epoch+1,'train_loss':avg_train_loss,'train_acc':train_acc,'val_loss':avg_val_loss,'val_acc':val_acc}
        wandb.log(epoch_metrics); print(f"Epoch {epoch+1:03d}: Train Acc: {train_acc:.4f}, Val Acc: {val_acc:.4f} | Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}")
        scheduler.step(); early_stopping(avg_val_loss, model)
        if early_stopping.early_stop: print("Early stopping"); break
    model.load_state_dict(torch.load(os.path.join(SAVE_DIR, 'react_wrn_best.pth'))); return model

model_path = os.path.join(SAVE_DIR, 'react_wrn_best.pth')
model = get_model()
if os.path.exists(model_path):
    print(f"--- Tải mô hình đã huấn luyện từ: {model_path} ---"); model.load_state_dict(torch.load(model_path, map_location=DEVICE))
else:
    print(f"--- Bắt đầu huấn luyện mô hình ReAct-WRN ---")
    # [SỬA LỖI QUAN TRỌNG] Truyền vào id_test_loader thay vì id_test_dataset
    model = train_model(model, id_train_loader, id_test_loader, epochs=200)

#==================================================================
# BƯỚC 4: ĐÁNH GIÁ
#==================================================================
print("\n" + " BƯỚC 4: ĐÁNH GIÁ ".center(80, "="))
model.calibrate(id_test_loader)
def evaluate_ood(model, id_loader, ood_loader):
    id_scores = []; ood_scores = []
    for images, _ in tqdm(id_loader, "Scoring ID data", leave=False): scores = model.ood_score(images.to(DEVICE)); id_scores.extend(scores)
    for images, _ in tqdm(ood_loader, "Scoring OOD data", leave=False): scores = model.ood_score(images.to(DEVICE)); ood_scores.extend(scores)
    labels = np.concatenate([np.zeros(len(id_scores)), np.ones(len(ood_scores))]); all_scores = np.concatenate([id_scores, ood_scores])
    auroc = roc_auc_score(labels, all_scores); fpr, tpr, _ = roc_curve(labels, all_scores); fpr95 = fpr[np.argmax(tpr >= 0.95)]
    return {'auroc': auroc, 'fpr95': fpr95}

results = []
print("\n--- Evaluating on Near-OOD (CIFAR non-animals) ---")
results.append({'Dataset': 'Near-OOD', **evaluate_ood(model, id_test_loader, near_ood_loader)})
print("--- Evaluating on Far-OOD (SVHN) ---")
results.append({'Dataset': 'Far-OOD', **evaluate_ood(model, id_test_loader, far_ood_loader)})
results_df = pd.DataFrame(results); print("\n" + " KẾT QUẢ ĐÁNH GIÁ ".center(80, "=")); print(results_df.to_string())

#==================================================================
# BƯỚC 5: WEB APPLICATION
#==================================================================
print("\n" + " BƯỚC 5: WEB APPLICATION ".center(80, "="))
class OODWebApp:
    def __init__(self, model, id_classes_names, test_loader):
        self.model, self.id_classes_names, self.transform = model, id_classes_names, transform_test
        print("Đang tính toán ngưỡng (threshold) cho Web App..."); id_scores = []
        for images, _ in tqdm(test_loader, "Calculating threshold", leave=False): id_scores.extend(self.model.ood_score(images.to(DEVICE)))
        self.threshold = np.percentile(id_scores, 95); print(f"Ngưỡng ReAct score cho Web App là: {self.threshold:.4f}")
    def predict(self, image):
        pil_image=Image.fromarray(image).convert("RGB"); image_tensor=self.transform(pil_image).unsqueeze(0).to(DEVICE); self.model.eval()
        ood_score = self.model.ood_score(image_tensor)[0]; is_ood = ood_score > self.threshold
        if is_ood: return {"status":"Out-of-Distribution", "ood_score":ood_score, "is_ood":True}
        else:
            with torch.no_grad(): logits = self.model(image_tensor)
            probs=F.softmax(logits,dim=1)[0]; pred_idx=torch.argmax(probs).item()
            return {"status":"In-Distribution", "ood_score":ood_score, "is_ood":False, "predicted_class":self.id_classes_names[pred_idx], "confidence":probs[pred_idx].item()}

web_app = OODWebApp(model, id_classes_names, id_test_loader)
def gradio_predict(image):
    result=web_app.predict(image); score_info=f"ReAct Score: {result['ood_score']:.4f} (Threshold: {web_app.threshold:.4f})"
    if result["is_ood"]: return f"🚨 **{result['status']}**","N/A","N/A",score_info
    else: return f"✅ **{result['status']}**",result["predicted_class"],f"{result['confidence']:.2%}",score_info

with gr.Blocks(title="🐾 OOD Animal Detection",theme=gr.themes.Soft()) as app:
    gr.Markdown("# 🐾 Out-of-Distribution Animal Detection\nUpload an image to detect if it's an animal or not, using an **Advanced ReAct model**.")
    with gr.Row():
        with gr.Column():
            image_input=gr.Image(label="📸 Upload Image",type="numpy"); predict_btn=gr.Button("🚀 Analyze Image",variant="primary")
        with gr.Column():
            status_output=gr.Markdown(label="📊 Detection Result"); class_output=gr.Textbox(label="🏷️ Predicted Animal Class"); confidence_output=gr.Textbox(label="🎯 Confidence"); score_output=gr.Textbox(label="📈 OOD Score Details")
    predict_btn.click(fn=gradio_predict,inputs=image_input,outputs=[status_output,class_output,confidence_output,score_output])

print("🚀 Launching Gradio Web Application...")
app.launch(share=True, debug=True)
```

```python
# -- coding: utf-8 --
"""
BÀI TOÁN NGOÀI PHÂN PHỐI ỨNG DỤNG TRONG NHẬN DIỆN MẪU
Phiên bản TỐI THƯỢNG: Contrastive Loss, Projection Head, AutoAugment, Advanced Training
"""
#==================================================================
# BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG
#==================================================================
print(" BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG ".center(80, "="))
!pip install -q torchmetrics gradio wandb diffusers transformers accelerate opencv-python scikit-image
!git clone -q https://github.com/deeplearning-wisc/dream-ood.git

import os, sys, torch, torchvision
sys.path.append('./dream-ood')
import torch.nn as nn, torch.optim as optim, torch.nn.functional as F
import torchvision.transforms as T
from torch.utils.data import DataLoader, Subset, ConcatDataset, Dataset
from tqdm.notebook import tqdm
import numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.metrics import roc_auc_score, roc_curve, auc
from PIL import Image
import gradio as gr, pandas as pd, wandb
from datetime import datetime

wandb.init(project="ood-animal-detection-ultimate", name=f"exp_{datetime.now().strftime('%Y%m%d_%H%M%S')}")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Sử dụng thiết bị: {DEVICE}")
from google.colab import drive
drive.mount('/content/drive')
SAVE_DIR = "/content/drive/MyDrive/OOD_Animal_Project_Ultimate" # Thư mục mới cho phiên bản tốt nhất
os.makedirs(SAVE_DIR, exist_ok=True)
print(f"Model và kết quả sẽ được lưu tại: {SAVE_DIR}")

#==================================================================
# BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU VỚI AUTOAUGMENT
#==================================================================
print("\n" + " BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU ".center(80, "="))
class DreamOODGenerator:
    def __init__(self, model_name="stabilityai/stable-diffusion-2-1-base"):
        from diffusers import StableDiffusionPipeline; self.pipe = StableDiffusionPipeline.from_pretrained(model_name, torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32)
        if torch.cuda.is_available(): self.pipe.to("cuda")
        self.ood_prompts = ["abstract geometric patterns","random noise texture","colorful abstract painting","geometric fractal patterns","digital glitch art"]
    def generate_ood_batch(self, batch_size=8, size=(32, 32)):
        images = []
        for _ in range(batch_size):
            prompt=f"{np.random.choice(['vibrant','muted','dark','pastel'])} {np.random.choice(self.ood_prompts)}"
            with torch.no_grad(): image = self.pipe(prompt, num_inference_steps=20, height=256, width=256, guidance_scale=7.5).images[0].resize(size)
            images.append(np.array(image))
        return np.array(images)
    def create_ood_dataset(self, num_samples=5000, save_path=None):
        print(f"Đang tạo {num_samples} mẫu OOD..."); batch_size=16; num_batches=(num_samples+batch_size-1)//batch_size
        all_images = np.concatenate([self.generate_ood_batch(min(batch_size, num_samples - i*batch_size)) for i in tqdm(range(num_batches), desc="Generating OOD")])
        if save_path: np.save(save_path, all_images); print(f"Đã lưu OOD dataset tại: {save_path}")
        return all_images
class DreamOODDataset(Dataset):
    def __init__(self, images, transform=None): self.images, self.transform = images, transform
    def __len__(self): return len(self.images)
    def __getitem__(self, idx):
        image = Image.fromarray(self.images[idx].astype(np.uint8));
        if self.transform: image = self.transform(image)
        return image, -1

normalize = T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
transform_train = T.Compose([
    T.RandomCrop(32, padding=4), T.RandomHorizontalFlip(),
    T.AutoAugment(T.AutoAugmentPolicy.CIFAR10),
    T.ToTensor(), normalize, T.RandomErasing()
])
transform_test = T.Compose([T.Resize(32), T.ToTensor(), normalize])
transform_oe = T.Compose([T.Resize((32, 32)), T.ToTensor(), T.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))])

dream_ood_path = os.path.join(SAVE_DIR, 'dream_ood_samples.npy')
if not os.path.exists(dream_ood_path):
    dream_generator = DreamOODGenerator(); dream_ood_images = dream_generator.create_ood_dataset(save_path=dream_ood_path)
else: print("Tải DreamOOD dataset đã có..."); dream_ood_images = np.load(dream_ood_path)
cifar10_classes=['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck'];cifar10_animal_indices=[2,3,4,5,6,7];cifar100_animal_indices=[1,3,4,6,7,14,15,18,19,21,26,27,30,31,32,34,36,38,42,43,44,45,46,51,56,57,65,66,67,68,70,71,72,73,74,75,76,77,86,91,93,98,99];cifar100_classes=['apple','aquarium_fish','baby','bear','beaver','bed','bee','beetle','bicycle','bottle','bowl','boy','bridge','bus','butterfly','camel','can','castle','caterpillar','cattle','chair','chimpanzee','clock','cloud','cockroach','couch','crab','crocodile','cup','dinosaur','dolphin','elephant','flatfish','forest','fox','girl','hamster','house','kangaroo','keyboard','lamp','lawn_mower','leopard','lion','lizard','lobster','man','maple_tree','motorcycle','mountain','mouse','mushroom','oak_tree','orange','orchid','otter','palm_tree','pear','pickup_truck','pine_tree','plain','plate','poppy','porcupine','possum','rabbit','raccoon','ray','road','rocket','rose','sea','seal','shark','shrew','skunk','skyscraper','snail','snake','spider','squirrel','streetcar','sunflower','sweet_pepper','table','tank','telephone','television','tiger','tractor','train','trout','tulip','turtle','wardrobe','whale','willow_tree','wolf','woman','worm']
id_classes_names=[cifar10_classes[i] for i in cifar10_animal_indices]+[cifar100_classes[i] for i in cifar100_animal_indices];near_ood_classes_indices=[i for i in range(10) if i not in cifar10_animal_indices]
class RemappedSubset(Subset):
    def __init__(self,d,i,m):super().__init__(d,i);self.label_map=m
    def __getitem__(self,idx):im,l=self.dataset[self.indices[idx]];return im,self.label_map[l]
cifar10_train=torchvision.datasets.CIFAR10('./data',train=True,download=True,transform=transform_train);cifar10_test=torchvision.datasets.CIFAR10('./data',train=False,download=True,transform=transform_test);cifar100_train=torchvision.datasets.CIFAR100('./data',train=True,download=True,transform=transform_train);cifar100_test=torchvision.datasets.CIFAR100('./data',train=False,download=True,transform=transform_test)
c10_label_map={o:n for n,o in enumerate(cifar10_animal_indices)};c100_label_map={o:n+len(c10_label_map) for n,o in enumerate(cifar100_animal_indices)}
id_train_dataset=ConcatDataset([RemappedSubset(cifar10_train,[i for i,(_,l)in enumerate(cifar10_train)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_train,[i for i,(_,l)in enumerate(cifar100_train)if l in cifar100_animal_indices],c100_label_map)])
id_test_dataset=ConcatDataset([RemappedSubset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_test,[i for i,(_,l)in enumerate(cifar100_test)if l in cifar100_animal_indices],c100_label_map)])
near_ood_test_dataset=Subset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in near_ood_classes_indices]);far_ood_test_dataset=torchvision.datasets.SVHN('./data',split='test',download=True,transform=transform_test);dream_ood_dataset=DreamOODDataset(dream_ood_images,transform=transform_oe)
batch_size=128;id_train_loader=DataLoader(id_train_dataset,batch_size,shuffle=True,num_workers=2,pin_memory=True);id_test_loader=DataLoader(id_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);near_ood_loader=DataLoader(near_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);far_ood_loader=DataLoader(far_ood_test_dataset,batch_size,shuffle=False,num_workers=2,pin_memory=True);dream_ood_loader=DataLoader(dream_ood_dataset,batch_size,shuffle=True,num_workers=2,pin_memory=True)

#==================================================================
# BƯỚC 3: KIẾN TRÚC MODEL TỐI THƯỢNG VÀ LOSS TƯƠNG PHẢN
#==================================================================
print("\n" + " BƯỚC 3: KIẾN TRÚC MODEL TỐI THƯỢNG ".center(80, "="))
class ProjectionHead(nn.Module):
    def __init__(self, in_features, projection_dim):
        super().__init__()
        self.projection = nn.Sequential(nn.Linear(in_features, in_features, bias=False), nn.ReLU(), nn.Linear(in_features, projection_dim, bias=False))
    def forward(self, x): return F.normalize(self.projection(x), p=2, dim=1)

class CustomResNet18(nn.Module):
    def __init__(self, num_classes, projection_dim=128):
        super().__init__()
        self.backbone = torchvision.models.resnet18(weights=torchvision.models.ResNet18_Weights.DEFAULT)
        self.backbone_dim = self.backbone.fc.in_features
        self.backbone.fc = nn.Identity()
        self.projection_head = ProjectionHead(self.backbone_dim, projection_dim)
        self.classifier = nn.Linear(self.backbone_dim, num_classes)
    def forward(self, x, return_features=False):
        backbone_features = self.backbone(x); logits = self.classifier(backbone_features)
        if return_features: projected_features = self.projection_head(backbone_features); return logits, backbone_features, projected_features
        return logits

class ContrastiveSeparationLoss(nn.Module):
    def __init__(self, margin=1.0):
        super().__init__(); self.margin = margin
    def forward(self, id_features, ood_features):
        loss_id = torch.mean(torch.norm(id_features, p=2, dim=1))
        loss_ood = F.relu(self.margin - torch.norm(ood_features, p=2, dim=1)).mean()
        return loss_id + loss_ood

def get_model(): return CustomResNet18(num_classes=len(id_classes_names)).to(DEVICE)
class OODDetector:
    @staticmethod
    def energy_score(model, data_loader):
        model.eval(); scores = [];
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "Energy Score", leave=False):
                logits = model(images.to(DEVICE)); energy = -torch.logsumexp(logits, dim=1)
                scores.extend(energy.cpu().numpy())
        return np.array(scores)
    @staticmethod
    def max_softmax_probability(model, data_loader):
        model.eval(); scores = []
        with torch.no_grad():
            for images, _ in tqdm(data_loader, "MSP", leave=False):
                probs = F.softmax(model(images.to(DEVICE)), dim=1); max_probs, _ = torch.max(probs, dim=1)
                scores.extend((-max_probs).cpu().numpy())
        return np.array(scores)
    class Mahalanobis:
        def fit(self, model, data_loader, num_classes):
            print("Fitting Mahalanobis detector..."); model.eval(); feature_dim = model.backbone_dim
            print(f"Detected feature dimension: {feature_dim}")
            self.class_means = torch.zeros(num_classes, feature_dim).to(DEVICE); class_counts = torch.zeros(num_classes).to(DEVICE); all_features = []
            with torch.no_grad():
                for images, labels in tqdm(data_loader, "Calculating means", leave=False):
                    images, labels = images.to(DEVICE), labels.to(DEVICE); _, backbone_features, _ = model(images, return_features=True); all_features.append(backbone_features)
                    for i in range(num_classes):
                        class_features = backbone_features[labels == i]
                        if len(class_features) > 0: self.class_means[i] += class_features.sum(0); class_counts[i] += len(class_features)
            class_counts[class_counts == 0] = 1; self.class_means /= class_counts.unsqueeze(1)
            all_features = torch.cat(all_features); all_labels = torch.cat([l for _, l in data_loader]).to(DEVICE)
            centered_features = all_features - self.class_means[all_labels]; covariance = torch.cov(centered_features.T)
            reg_covariance = covariance + torch.eye(feature_dim).to(DEVICE) * 1e-6
            self.precision = torch.linalg.inv(reg_covariance); print("Mahalanobis fit complete.")
        def score(self, model, data_loader):
            print("Scoring with Mahalanobis..."); model.eval(); scores = []
            with torch.no_grad():
                for images, _ in tqdm(data_loader, "Mahalanobis Score", leave=False):
                    _, backbone_features, _ = model(images.to(DEVICE), return_features=True); min_dists = []
                    for feature in backbone_features:
                        dists = [torch.sqrt((feature - mean) @ self.precision @ (feature - mean).T) for mean in self.class_means]
                        min_dists.append(torch.min(torch.stack(dists)))
                    scores.extend(torch.stack(min_dists).cpu().numpy())
            return np.array(scores)
def evaluate_ood_performance(id_scores, ood_scores):
    labels=np.concatenate([np.zeros_like(id_scores),np.ones_like(ood_scores)]);scores=np.concatenate([id_scores,ood_scores])
    auroc=roc_auc_score(labels,scores);fpr,tpr,_=roc_curve(labels,scores);fpr95=fpr[np.argmax(tpr>=0.95)]
    return {'auroc':auroc,'fpr95':fpr95}

#==================================================================
# BƯỚC 4: QUY TRÌNH HUẤN LUYỆN TỐI THƯỢNG
#==================================================================
print("\n" + " BƯỚC 4: QUY TRÌNH HUẤN LUYỆN TỐI THƯỢNG ".center(80, "="))
class EarlyStopping:
    def __init__(self, patience=7, verbose=False, delta=0, path='checkpoint.pt'):
        self.patience, self.verbose, self.delta, self.path = patience, verbose, delta, path; self.counter, self.best_score, self.early_stop = 0, None, False; self.val_loss_min = np.inf
    def __call__(self, val_loss, model):
        score = -val_loss
        if self.best_score is None: self.best_score = score; self.save_checkpoint(val_loss, model)
        elif score < self.best_score + self.delta:
            self.counter += 1; print(f'EarlyStopping counter: {self.counter} out of {self.patience}')
            if self.counter >= self.patience: self.early_stop = True
        else: self.best_score = score; self.save_checkpoint(val_loss, model); self.counter = 0
    def save_checkpoint(self, val_loss, model):
        if self.verbose: print(f'Validation loss decreased ({self.val_loss_min:.6f} --> {val_loss:.6f}).  Saving model ...')
        torch.save(model.state_dict(), self.path); self.val_loss_min = val_loss

def train_ultimate_model(model, train_loader, val_loader, ood_loader, epochs=100):
    optimizer = optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-2)
    # [SỬA LỖI QUAN TRỌNG] Xóa bỏ `verbose=True` để tương thích PyTorch mới
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=5, factor=0.2)
    early_stopping = EarlyStopping(patience=15, verbose=True, path=os.path.join(SAVE_DIR, 'ultimate_model.pth'))
    cls_criterion = nn.CrossEntropyLoss(); contrastive_criterion = ContrastiveSeparationLoss(margin=1.0); contrastive_weight = 1.0
    for epoch in range(epochs):
        model.train(); train_loss=0; train_correct=0; train_total=0; ood_iter = iter(ood_loader)
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}", leave=False)
        for id_inputs, id_targets in progress_bar:
            id_inputs, id_targets = id_inputs.to(DEVICE), id_targets.to(DEVICE)
            try: ood_inputs, _ = next(ood_iter)
            except StopIteration: ood_iter = iter(ood_loader); ood_inputs, _ = next(ood_iter)
            ood_inputs = ood_inputs.to(DEVICE); optimizer.zero_grad()
            all_inputs = torch.cat([id_inputs, ood_inputs]); logits, _, projected_features = model(all_inputs, return_features=True)
            id_logits = logits[:len(id_inputs)]; id_projected_features = projected_features[:len(id_inputs)]; ood_projected_features = projected_features[len(id_inputs):]
            loss_cls = cls_criterion(id_logits, id_targets); loss_contrastive = contrastive_criterion(id_projected_features, ood_projected_features)
            total_loss = loss_cls + contrastive_weight * loss_contrastive
            total_loss.backward(); torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0); optimizer.step()
            train_loss+=total_loss.item(); _,predicted=id_logits.max(1); train_total+=id_targets.size(0); train_correct+=predicted.eq(id_targets).sum().item()
        model.eval(); val_loss=0; val_correct=0; val_total=0
        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs,targets=inputs.to(DEVICE),targets.to(DEVICE); outputs=model(inputs); loss=cls_criterion(outputs,targets)
                val_loss+=loss.item(); _,predicted=outputs.max(1); val_total+=targets.size(0); val_correct+=predicted.eq(targets).sum().item()
        avg_train_loss=train_loss/len(train_loader); avg_val_loss=val_loss/len(val_loader); train_acc,val_acc=train_correct/train_total,val_correct/val_total
        epoch_metrics={'epoch':epoch+1,'train_loss':avg_train_loss,'train_acc':train_acc,'val_loss':avg_val_loss,'val_acc':val_acc}
        wandb.log({f'ultimate/{k}':v for k,v in epoch_metrics.items()})
        print(f"Epoch {epoch+1:03d}: Train Acc: {train_acc:.4f}, Val Acc: {val_acc:.4f} | Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}")
        scheduler.step(avg_val_loss); early_stopping(avg_val_loss, model)
        if early_stopping.early_stop: print("Early stopping"); break
    model.load_state_dict(torch.load(os.path.join(SAVE_DIR, 'ultimate_model.pth'))); return model

model_path = os.path.join(SAVE_DIR, 'ultimate_model.pth')
ultimate_model = get_model()
if os.path.exists(model_path):
    print(f"--- Tải mô hình đã huấn luyện từ: {model_path} ---"); ultimate_model.load_state_dict(torch.load(model_path, map_location=DEVICE))
else:
    print(f"--- Bắt đầu huấn luyện mô hình TỐI THƯỢNG ---"); ultimate_model = train_ultimate_model(ultimate_model, id_train_loader, id_test_loader, dream_ood_loader, epochs=100)

#==================================================================
# BƯỚC 5 & 6: ĐÁNH GIÁ VÀ TRIỂN KHAI
#==================================================================
print("\n" + " BƯỚC 5 & 6: ĐÁNH GIÁ VÀ TRIỂN KHAI ".center(80, "="))
mahalanobis_detector = OODDetector.Mahalanobis(); mahalanobis_detector.fit(ultimate_model, id_train_loader, len(id_classes_names))
ood_methods = {'Energy': OODDetector.energy_score, 'MSP': OODDetector.max_softmax_probability, 'Mahalanobis': mahalanobis_detector.score}
results = []
for name, func in ood_methods.items():
    id_scores = func(ultimate_model, id_test_loader); near_scores = func(ultimate_model, near_ood_loader); far_scores = func(ultimate_model, far_ood_loader)
    results.append({'Method':name,'Dataset':'Near-OOD',**evaluate_ood_performance(id_scores,near_scores)}); results.append({'Method':name,'Dataset':'Far-OOD',**evaluate_ood_performance(id_scores,far_scores)})
results_df = pd.DataFrame(results); print("\n" + " KẾT QUẢ ĐÁNH GIÁ ".center(80, "=")); print(results_df.to_string())
fig, axes = plt.subplots(1, 2, figsize=(18, 7)); plt.suptitle('Hiệu suất Phát hiện OOD của Model Tối thượng', fontsize=16)
for i, dataset in enumerate(['Near-OOD', 'Far-OOD']):
    ax = axes[i]; data = results_df[results_df['Dataset'] == dataset]; sns.barplot(x='Method', y='auroc', data=data, ax=ax)
    ax.set_title(f'AUROC on {dataset}'); ax.set_ylabel('AUROC Score'); ax.set_ylim(0.5, 1.0)
    for p in ax.patches: ax.annotate(f'{p.get_height():.4f}',(p.get_x()+p.get_width()/2.,p.get_height()),ha='center',va='center',xytext=(0,5),textcoords='offset points')
plt.tight_layout(rect=[0,0,1,0.95]); plt.show(); wandb.log({"Ultimate OOD Performance": wandb.Image(plt)})

class OODWebApp:
    def __init__(self, model, id_classes_names, test_loader):
        self.model, self.id_classes_names, self.transform = model, id_classes_names, transform_test
        print("Đang tính toán ngưỡng (threshold)..."); id_scores = OODDetector.energy_score(self.model, test_loader)
        self.threshold = np.percentile(id_scores, 95); print(f"Ngưỡng Energy score được xác định là: {self.threshold:.4f}")
    def predict(self, image):
        pil_image=Image.fromarray(image).convert("RGB"); image_tensor=self.transform(pil_image).unsqueeze(0).to(DEVICE); self.model.eval()
        with torch.no_grad():
            logits=self.model(image_tensor); ood_score=-torch.logsumexp(logits,dim=1).item(); is_ood=ood_score>self.threshold
            if is_ood: return {"status":"Out-of-Distribution","ood_score":ood_score,"is_ood":True}
            else:
                probs=F.softmax(logits,dim=1)[0]; pred_idx=torch.argmax(probs).item()
                return {"status":"In-Distribution","ood_score":ood_score,"is_ood":False,"predicted_class":self.id_classes_names[pred_idx],"confidence":probs[pred_idx].item()}

web_app = OODWebApp(ultimate_model, id_classes_names, id_test_loader)
def gradio_predict(image):
    result=web_app.predict(image); score_info=f"Energy Score: {result['ood_score']:.4f} (Threshold: {web_app.threshold:.4f})"
    if result["is_ood"]: return f"🚨 **{result['status']}**","N/A","N/A",score_info
    else: return f"✅ **{result['status']}**",result["predicted_class"],f"{result['confidence']:.2%}",score_info

with gr.Blocks(title="🐾 OOD Animal Detection",theme=gr.themes.Soft()) as app:
    gr.Markdown("# 🐾 Out-of-Distribution Animal Detection\nUpload an image to detect if it's an animal or not, using the **Ultimate DreamOE model**.")
    with gr.Row():
        with gr.Column():
            image_input=gr.Image(label="📸 Upload Image",type="numpy"); predict_btn=gr.Button("🚀 Analyze Image",variant="primary")
        with gr.Column():
            status_output=gr.Markdown(label="📊 Detection Result"); class_output=gr.Textbox(label="🏷️ Predicted Animal Class"); confidence_output=gr.Textbox(label="🎯 Confidence"); score_output=gr.Textbox(label="📈 OOD Score Details")
    predict_btn.click(fn=gradio_predict,inputs=image_input,outputs=[status_output,class_output,confidence_output,score_output])

print("🚀 Launching Gradio Web Application...")
app.launch(share=True, debug=True)
```

```python
# -- coding: utf-8 --
"""
BÀI TOÁN NGOÀI PHÂN PHỐI ỨNG DỤNG TRONG NHẬN DIỆN MẪU
Phiên bản PAPER-INSPIRED: ResNet-18 + Margin-based Contrastive Loss + Mahalanobis Distance
"""
#==================================================================
# BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG
#==================================================================
print(" BƯỚC 0: CÀI ĐẶT MÔI TRƯỜNG ".center(80, "="))
!pip install -q torchmetrics gradio wandb diffusers transformers accelerate opencv-python scikit-image
!git clone -q https://github.com/deeplearning-wisc/dream-ood.git

import os, sys, torch, torchvision
sys.path.append('./dream-ood')
import torch.nn as nn, torch.optim as optim, torch.nn.functional as F
import torchvision.transforms as T
from torch.utils.data import DataLoader, Subset, ConcatDataset, Dataset
from tqdm.notebook import tqdm
import numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.metrics import roc_auc_score, roc_curve, auc
from PIL import Image
import gradio as gr, pandas as pd, wandb
from datetime import datetime

wandb.init(project="ood-animal-detection-paper-inspired", name=f"exp_{datetime.now().strftime('%Y%m%d_%H%M%S')}")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Sử dụng thiết bị: {DEVICE}")
from google.colab import drive
drive.mount('/content/drive')
SAVE_DIR = "/content/drive/MyDrive/OOD_Animal_Project_Final_Versionwithpaper"
os.makedirs(SAVE_DIR, exist_ok=True)
print(f"Model và kết quả sẽ được lưu tại: {SAVE_DIR}")

#==================================================================
# BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU
#==================================================================
print("\n" + " BƯỚC 1 & 2: CHUẨN BỊ DỮ LIỆU ".center(80, "="))
class DreamOODGenerator:
    def __init__(self, model_name="stabilityai/stable-diffusion-2-1-base"):
        from diffusers import StableDiffusionPipeline; self.pipe = StableDiffusionPipeline.from_pretrained(model_name, torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32)
        if torch.cuda.is_available(): self.pipe.to("cuda")
        self.ood_prompts = ["abstract geometric patterns","random noise texture","colorful abstract painting","geometric fractal patterns","digital glitch art"]
    def generate_ood_batch(self, batch_size=8, size=(32, 32)):
        images = []
        for _ in range(batch_size):
            prompt=f"{np.random.choice(['vibrant','muted','dark','pastel'])} {np.random.choice(self.ood_prompts)}"
            with torch.no_grad(): image = self.pipe(prompt, num_inference_steps=20, height=256, width=256, guidance_scale=7.5).images[0].resize(size)
            images.append(np.array(image))
        return np.array(images)
    def create_ood_dataset(self, num_samples=5000, save_path=None):
        print(f"Đang tạo {num_samples} mẫu OOD..."); batch_size=16; num_batches=(num_samples+batch_size-1)//batch_size
        all_images = np.concatenate([self.generate_ood_batch(min(batch_size, num_samples - i*batch_size)) for i in tqdm(range(num_batches), desc="Generating OOD")])
        if save_path: np.save(save_path, all_images); print(f"Đã lưu OOD dataset tại: {save_path}")
        return all_images
class DreamOODDataset(Dataset):
    def __init__(self, images, transform=None): self.images, self.transform = images, transform
    def __len__(self): return len(self.images)
    def __getitem__(self, idx):
        image = Image.fromarray(self.images[idx].astype(np.uint8));
        if self.transform: image = self.transform(image)
        return image, -1

normalize = T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
transform_train = T.Compose([T.RandomCrop(32, padding=4), T.RandomHorizontalFlip(), T.AutoAugment(T.AutoAugmentPolicy.CIFAR10), T.ToTensor(), normalize, T.RandomErasing()])
transform_test = T.Compose([T.Resize(32), T.ToTensor(), normalize])
transform_oe = T.Compose([T.Resize((32, 32)), T.ToTensor(), T.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))])
dream_ood_path = os.path.join(SAVE_DIR, 'dream_ood_samples.npy')
if not os.path.exists(dream_ood_path):
    dream_generator = DreamOODGenerator(); dream_ood_images = dream_generator.create_ood_dataset(save_path=dream_ood_path)
else: print("Tải DreamOOD dataset đã có..."); dream_ood_images = np.load(dream_ood_path)
cifar10_classes=['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck'];cifar10_animal_indices=[2,3,4,5,6,7];cifar100_animal_indices=[1,3,4,6,7,14,15,18,19,21,26,27,30,31,32,34,36,38,42,43,44,45,46,51,56,57,65,66,67,68,70,71,72,73,74,75,76,77,86,91,93,98,99];cifar100_classes=['apple','aquarium_fish','baby','bear','beaver','bed','bee','beetle','bicycle','bottle','bowl','boy','bridge','bus','butterfly','camel','can','castle','caterpillar','cattle','chair','chimpanzee','clock','cloud','cockroach','couch','crab','crocodile','cup','dinosaur','dolphin','elephant','flatfish','forest','fox','girl','hamster','house','kangaroo','keyboard','lamp','lawn_mower','leopard','lion','lizard','lobster','man','maple_tree','motorcycle','mountain','mouse','mushroom','oak_tree','orange','orchid','otter','palm_tree','pear','pickup_truck','pine_tree','plain','plate','poppy','porcupine','possum','rabbit','raccoon','ray','road','rocket','rose','sea','seal','shark','shrew','skunk','skyscraper','snail','snake','spider','squirrel','streetcar','sunflower','sweet_pepper','table','tank','telephone','television','tiger','tractor','train','trout','tulip','turtle','wardrobe','whale','willow_tree','wolf','woman','worm']
id_classes_names=[cifar10_classes[i] for i in cifar10_animal_indices]+[cifar100_classes[i] for i in cifar100_animal_indices];near_ood_classes_indices=[i for i in range(10) if i not in cifar10_animal_indices]
class RemappedSubset(Subset):
    def __init__(self,d,i,m):super().__init__(d,i);self.label_map=m
    def __getitem__(self,idx):im,l=self.dataset[self.indices[idx]];return im,self.label_map[l]
cifar10_train=torchvision.datasets.CIFAR10('./data',train=True,download=True,transform=transform_train);cifar10_test=torchvision.datasets.CIFAR10('./data',train=False,download=True,transform=transform_test);cifar100_train=torchvision.datasets.CIFAR100('./data',train=True,download=True,transform=transform_train);cifar100_test=torchvision.datasets.CIFAR100('./data',train=False,download=True,transform=transform_test)
c10_label_map={o:n for n,o in enumerate(cifar10_animal_indices)};c100_label_map={o:n+len(c10_label_map) for n,o in enumerate(cifar100_animal_indices)}
id_train_dataset=ConcatDataset([RemappedSubset(cifar10_train,[i for i,(_,l)in enumerate(cifar10_train)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_train,[i for i,(_,l)in enumerate(cifar100_train)if l in cifar100_animal_indices],c100_label_map)])
id_test_dataset=ConcatDataset([RemappedSubset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_test,[i for i,(_,l)in enumerate(cifar100_test)if l in cifar100_animal_indices],c100_label_map)])
near_ood_test_dataset=Subset(cifar10_test,[i for i,(_,l)in enumerate(cifar10_test)if l in near_ood_classes_indices]);far_ood_test_dataset=torchvision.datasets.SVHN('./data',split='test',download=True,transform=transform_test);dream_ood_dataset=DreamOODDataset(dream_ood_images,transform=transform_oe)
batch_size=128;id_train_loader=DataLoader(id_train_dataset,batch_size,shuffle=True,num_workers=0);id_test_loader=DataLoader(id_test_dataset,batch_size,shuffle=False,num_workers=0);near_ood_loader=DataLoader(near_ood_test_dataset,batch_size,shuffle=False,num_workers=0);far_ood_loader=DataLoader(far_ood_test_dataset,batch_size,shuffle=False,num_workers=0);dream_ood_loader=DataLoader(dream_ood_dataset,batch_size,shuffle=True,num_workers=0)

#==================================================================
# BƯỚC 3: KIẾN TRÚC MODEL VÀ LOSS
#==================================================================
print("\n" + " BƯỚC 3: KIẾN TRÚC MODEL VÀ LOSS ".center(80, "="))
class FinalResNet18(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.backbone = torchvision.models.resnet18(weights=torchvision.models.ResNet18_Weights.DEFAULT)
        self.backbone_dim = self.backbone.fc.in_features
        self.backbone.fc = nn.Identity()
        self.classifier = nn.Linear(self.backbone_dim, num_classes)
    def forward(self, x, return_features=False):
        features = self.backbone(x); logits = self.classifier(features)
        if return_features: return logits, features
        return logits

# [CẢI TIẾN VƯỢT TRỘI] Margin-based Contrastive Loss từ bài báo
class MarginContrastiveLoss(nn.Module):
    def forward(self, features, labels):
        # Tính toán pairwise L2 distance
        pairwise_dist = torch.cdist(features, features, p=2)
        # Tạo mask cho các cặp positive (cùng lớp) và negative (khác lớp)
        bsz = labels.shape[0]
        mask_positive = (labels.unsqueeze(0) == labels.unsqueeze(1)).float()
        mask_negative = 1 - mask_positive
        mask_diag = torch.eye(bsz, device=DEVICE)
        mask_positive.masked_fill_(mask_diag.bool(), 0) # Loại bỏ cặp (i, i)

        # Loss cho các cặp positive: kéo chúng lại gần nhau
        # Chỉ tính loss nếu có cặp positive
        if mask_positive.sum() > 0:
            loss_pos = (pairwise_dist * mask_positive).sum() / (mask_positive.sum() + 1e-8)
        else:
            loss_pos = 0.0

        # Margin thích ứng: là khoảng cách lớn nhất giữa các cặp positive trong batch
        if mask_positive.sum() > 0:
            adaptive_margin = (pairwise_dist * mask_positive).max()
        else:
            adaptive_margin = 0.0

        # Loss cho các cặp negative: đẩy chúng ra xa nhau một khoảng margin
        loss_neg = F.relu(adaptive_margin.detach() - pairwise_dist) * mask_negative
        if mask_negative.sum() > 0:
            loss_neg = loss_neg.sum() / (mask_negative.sum() + 1e-8)
        else:
            loss_neg = 0.0

        return loss_pos + loss_neg

def get_model(): return FinalResNet18(num_classes=len(id_classes_names)).to(DEVICE)
class OODDetector:
    # [CẢI TIẾN VƯỢT TRỘI] Mahalanobis sẽ là phương pháp chính
    class Mahalanobis:
        def fit(self, model, data_loader, num_classes):
            print("Fitting Mahalanobis detector..."); model.eval(); feature_dim = model.backbone_dim
            self.class_means = torch.zeros(num_classes, feature_dim).to(DEVICE); class_counts = torch.zeros(num_classes).to(DEVICE)
            with torch.no_grad():
                for images, labels in tqdm(data_loader, "Calculating class means", leave=False):
                    images, labels = images.to(DEVICE), labels.to(DEVICE); _, features = model(images, return_features=True)
                    for i in range(num_classes):
                        class_features = features[labels == i]
                        if len(class_features) > 0: self.class_means[i] += class_features.sum(0); class_counts[i] += len(class_features)
            class_counts[class_counts == 0] = 1; self.class_means /= class_counts.unsqueeze(1)
            all_features = []; all_labels_list = []
            with torch.no_grad():
                for images, labels in data_loader:
                    _, features = model(images.to(DEVICE), return_features=True)
                    all_features.append(features.cpu()); all_labels_list.append(labels)
            all_features = torch.cat(all_features); all_labels = torch.cat(all_labels_list)
            centered_features = all_features - self.class_means[all_labels].cpu()
            covariance = torch.cov(centered_features.T)
            reg_covariance = covariance + torch.eye(feature_dim) * 1e-3 # Tăng regularization
            self.precision = torch.linalg.inv(reg_covariance).to(DEVICE); print("Fit complete.")
        def score(self, model, data_loader):
            model.eval(); all_scores = []
            with torch.no_grad():
                for images, _ in tqdm(data_loader, "Scoring with Mahalanobis", leave=False):
                    _, features = model(images.to(DEVICE), return_features=True)
                    diffs = features.unsqueeze(1) - self.class_means.unsqueeze(0) # (B, C, D)
                    dists = torch.sqrt(torch.einsum('bcd,de,bce->bc', diffs, self.precision, diffs))
                    min_dists, _ = torch.min(dists, dim=1)
                    all_scores.extend(min_dists.cpu().numpy())
            return np.array(all_scores)
def evaluate_ood_performance(id_scores, ood_scores):
    labels=np.concatenate([np.zeros_like(id_scores),np.ones_like(ood_scores)]);scores=np.concatenate([id_scores,ood_scores])
    auroc=roc_auc_score(labels,scores);fpr,tpr,_=roc_curve(labels,scores);fpr95=fpr[np.argmax(tpr>=0.95)]
    return {'auroc':auroc,'fpr95':fpr95}

#==================================================================
# BƯỚC 4: QUY TRÌNH HUẤN LUYỆN NÂNG CAO
#==================================================================
print("\n" + " BƯỚC 4: QUY TRÌNH HUẤN LUYỆN NÂNG CAO ".center(80, "="))
class EarlyStopping:
    def __init__(self, patience=7, verbose=True, delta=0, path='checkpoint.pt'):
        self.patience,self.verbose,self.delta,self.path = patience,verbose,delta,path;self.counter,self.best_score,self.early_stop=0,None,False;self.val_loss_min=np.inf
    def __call__(self, val_loss, model):
        if val_loss < self.val_loss_min - self.delta: self.save_checkpoint(val_loss, model); self.val_loss_min = val_loss; self.counter = 0
        else:
            self.counter += 1; print(f'EarlyStopping counter: {self.counter} out of {self.patience}')
            if self.counter >= self.patience: self.early_stop = True
    def save_checkpoint(self, val_loss, model):
        if self.verbose: print(f'Validation loss decreased ({self.val_loss_min:.6f} --> {val_loss:.6f}).  Saving model ...')
        torch.save(model.state_dict(), self.path)

def train_final_model(model, train_loader, val_loader, epochs=100):
    optimizer = optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-2)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)
    early_stopping = EarlyStopping(patience=20, verbose=True, path=os.path.join(SAVE_DIR, 'final_model.pth'))
    cls_criterion = nn.CrossEntropyLoss(); margin_criterion = MarginContrastiveLoss()
    margin_weight = 0.1 # Trọng số cho margin loss

    for epoch in range(epochs):
        model.train(); train_loss=0; train_correct=0; train_total=0
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}", leave=False)
        for id_inputs, id_targets in progress_bar:
            id_inputs, id_targets = id_inputs.to(DEVICE), id_targets.to(DEVICE); optimizer.zero_grad()
            id_logits, id_features = model(id_inputs, return_features=True)
            loss_cls = cls_criterion(id_logits, id_targets)
            loss_margin = margin_criterion(id_features, id_targets)
            total_loss = loss_cls + margin_weight * loss_margin
            total_loss.backward(); optimizer.step()
            train_loss+=total_loss.item(); _,predicted=id_logits.max(1); train_total+=id_targets.size(0); train_correct+=predicted.eq(id_targets).sum().item()
        model.eval(); val_loss=0; val_correct=0; val_total=0
        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs,targets=inputs.to(DEVICE),targets.to(DEVICE); outputs=model(inputs); loss=cls_criterion(outputs,targets)
                val_loss+=loss.item(); _,predicted=outputs.max(1); val_total+=targets.size(0); val_correct+=predicted.eq(targets).sum().item()
        avg_train_loss=train_loss/len(train_loader); avg_val_loss=val_loss/len(val_loader); train_acc,val_acc=train_correct/train_total,val_correct/val_total
        print(f"Epoch {epoch+1:03d}: Train Acc: {train_acc:.4f}, Val Acc: {val_acc:.4f} | Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}")
        scheduler.step(); early_stopping(avg_val_loss, model)
        if early_stopping.early_stop: print("Early stopping"); break
    model.load_state_dict(torch.load(os.path.join(SAVE_DIR, 'final_model.pth'))); return model

model_path = os.path.join(SAVE_DIR, 'final_model.pth')
final_model = get_model()
if os.path.exists(model_path):
    print(f"--- Tải mô hình đã huấn luyện từ: {model_path} ---"); final_model.load_state_dict(torch.load(model_path, map_location=DEVICE))
else:
    print(f"--- Bắt đầu huấn luyện mô hình Final ---"); final_model = train_final_model(final_model, id_train_loader, id_test_loader, epochs=100)

#==================================================================
# BƯỚC 5 & 6: ĐÁNH GIÁ VÀ TRIỂN KHAI
#==================================================================
print("\n" + " BƯỚC 5 & 6: ĐÁNH GIÁ VÀ TRIỂN KHAI ".center(80, "="))
id_train_loader_for_fitting = DataLoader(ConcatDataset([RemappedSubset(cifar10_train,[i for i,(_,l)in enumerate(cifar10_train)if l in cifar10_animal_indices],c10_label_map),RemappedSubset(cifar100_train,[i for i,(_,l)in enumerate(cifar100_train)if l in cifar100_animal_indices],c100_label_map)]),batch_size=batch_size,shuffle=False,num_workers=0)
maha_detector = OODDetector.Mahalanobis(); maha_detector.fit(final_model, id_train_loader_for_fitting, len(id_classes_names))
ood_methods = {'Mahalanobis': maha_detector.score}
results = []
for name, func in ood_methods.items():
    id_scores = func(final_model, id_test_loader); near_scores = func(final_model, near_ood_loader); far_scores = func(final_model, far_ood_loader)
    results.append({'Method':name,'Dataset':'Near-OOD',**evaluate_ood_performance(id_scores,near_scores)}); results.append({'Method':name,'Dataset':'Far-OOD',**evaluate_ood_performance(id_scores,far_scores)})
results_df = pd.DataFrame(results); print("\n" + " KẾT QUẢ ĐÁNH GIÁ ".center(80, "=")); print(results_df.to_string())
# (Phần vẽ biểu đồ có thể bỏ qua vì chỉ có 1 phương pháp)

class OODWebApp:
    def __init__(self, model, detector):
        self.model, self.detector = model, detector; self.id_classes_names, self.transform = id_classes_names, transform_test
        print("Đang tính toán ngưỡng (threshold)..."); id_scores = self.detector.score(self.model, id_test_loader)
        self.threshold = np.percentile(id_scores, 95); print(f"Ngưỡng Mahalanobis Distance được xác định là: {self.threshold:.4f}")
    def predict(self, image):
        pil_image=Image.fromarray(image).convert("RGB"); image_tensor=self.transform(pil_image).unsqueeze(0).to(DEVICE); self.model.eval()
        with torch.no_grad():
            logits, features = self.model(image_tensor, return_features=True)
            dists = torch.cdist(features, self.detector.class_means); ood_score, _ = torch.min(dists, dim=1); ood_score = ood_score.item()
            is_ood = ood_score > self.threshold
            if is_ood: return {"status":"Out-of-Distribution","ood_score":ood_score,"is_ood":True}
            else:
                probs=F.softmax(logits,dim=1)[0]; pred_idx=torch.argmax(probs).item()
                return {"status":"In-Distribution","ood_score":ood_score,"is_ood":False,"predicted_class":self.id_classes_names[pred_idx],"confidence":probs[pred_idx].item()}

web_app = OODWebApp(final_model, maha_detector)
def gradio_predict(image):
    result=web_app.predict(image); score_info=f"Mahalanobis Distance: {result['ood_score']:.4f} (Threshold: {web_app.threshold:.4f})"
    if result["is_ood"]: return f"🚨 **{result['status']}**","N/A","N/A",score_info
    else: return f"✅ **{result['status']}**",result["predicted_class"],f"{result['confidence']:.2%}",score_info

with gr.Blocks(title="🐾 OOD Animal Detection",theme=gr.themes.Soft()) as app:
    gr.Markdown("# 🐾 Out-of-Distribution Animal Detection\nUpload an image to detect if it's an animal or not, using the **Final model**.")
    with gr.Row():
        with gr.Column():
            image_input=gr.Image(label="📸 Upload Image",type="numpy"); predict_btn=gr.Button("🚀 Analyze Image",variant="primary")
        with gr.Column():
            status_output=gr.Markdown(label="📊 Detection Result"); class_output=gr.Textbox(label="🏷️ Predicted Animal Class"); confidence_output=gr.Textbox(label="🎯 Confidence"); score_output=gr.Textbox(label="📈 OOD Score Details")
    predict_btn.click(fn=gradio_predict,inputs=image_input,outputs=[status_output,class_output,confidence_output,score_output])

print("🚀 Launching Gradio Web Application...")
app.launch(share=True, debug=True)
```
